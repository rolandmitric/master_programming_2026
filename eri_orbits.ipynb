```python
from itertools import product, permutations

def triples(L):
    return [(i, j, k)
            for n in range(L + 1)
            for i in range(n + 1)
            for j in range(n + 1 - i)
            for k in [n - i - j]]

perms = list(permutations([0, 1, 2]))

def p3(t, perm):
    return tuple(t[i] for i in perm)

def eri_sym(q):
    a, b, c, d = q
    out = set()
    for ab in [(a, b), (b, a)]:
        for cd in [(c, d), (d, c)]:
            out.add(ab + cd)
            out.add(cd + ab)
    return out

def orbit(q):
    return {
        tuple(p3(t, perm) for t in qq)
        for qq in eri_sym(q)
        for perm in perms
    }

def canon(q):
    return min(orbit(q))

def minimal_eri_tuples(L):
    T = triples(L)
    return sorted({canon(q) for q in product(T, repeat=4)})

reps = minimal_eri_tuples(2)

print(len(reps))   # 313
for r in reps[:20]:
    print(r)
```
