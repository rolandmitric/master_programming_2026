# One-Body, Two-Body, and General Many-Body Operators in Second Quantization

## From particle statistics to physical observables

We have constructed fermionic Fock space from a vacuum state and creation operators,

$$
|\Phi\rangle
=
a_{i_1}^\dagger a_{i_2}^\dagger \cdots a_{i_{N_e}}^\dagger |0\rangle,
$$

with canonical anticommutation relations

$$
\{a_p,a_q^\dagger\}=\delta_{pq},
\qquad
\{a_p,a_q\}=0,
\qquad
\{a_p^\dagger,a_q^\dagger\}=0.
$$

The total particle-number operator is

$$
\hat{\mathcal N}
=
\sum_p a_p^\dagger a_p,
$$

and satisfies

$$
[\hat{\mathcal N},a_p^\dagger]=a_p^\dagger,
\qquad
[\hat{\mathcal N},a_p]=-a_p.
$$

We now ask the next fundamental question:

$$
\boxed{
\text{How are ordinary physical operators represented on Fock space?}
}
$$

For molecular electronic structure this means: how do the familiar one-electron and two-electron integrals of first quantization become operators acting on determinants and configuration-interaction states?

## Learning goals and notation

At the end of these notes, we will understand why

$$
\hat O^{(1)}
=
\sum_{pq} o_{pq}\,a_p^\dagger a_q
$$

represents a general one-body operator and why

$$
\hat W^{(2)}
=
\frac{1}{2}
\sum_{pqrs}
(pq|rs)\,
a_p^\dagger a_q^\dagger a_s a_r
$$

represents a general symmetric two-body interaction.

Throughout:

- \(p,q,r,s,\ldots\) label orthonormal **spin-orbitals** \(\chi_p(x)\).
- \(x=(\mathbf r,\omega)\) denotes spatial and spin coordinates together.
- We consider fermions, appropriate for electrons.
- The notation

$$
(pq|rs)
=
\iint dx_1\,dx_2\,
\chi_p^\ast(x_1)\chi_q^\ast(x_2)
\,\hat g(1,2)\,
\chi_r(x_1)\chi_s(x_2)
$$

is used for ordered two-particle spin-orbital matrix elements. This is sometimes called the *physicists' ordering* of two-electron integrals. Different quantum-chemistry texts and programs may reorder the printed indices; the defining integral is unambiguous.

## 1. Particle number and body rank are different concepts

An \(N_e\)-electron state contains \(N_e\) electrons. By contrast, the **body rank** of an operator tells us how many electrons the operator acts upon simultaneously.

A one-body operator acts on one electron at a time:

$$
\hat O^{(1)}_{N_e}
=
\sum_{\alpha=1}^{N_e}
\hat o(\alpha).
$$

A two-body operator acts on electron pairs:

$$
\hat W^{(2)}_{N_e}
=
\sum_{\alpha<\beta}
\hat g(\alpha,\beta).
$$

A three-body operator would act on triples:

$$
\hat U^{(3)}_{N_e}
=
\sum_{\alpha<\beta<\gamma}
\hat u(\alpha,\beta,\gamma).
$$

Thus, even though the physical state may contain many electrons, an electronic Hamiltonian with pairwise Coulomb repulsion contains only zero-, one-, and two-body operators.

$$
\boxed{
\text{Number of particles in the state}
\neq
\text{body rank of the operator}.
}
$$

## 2. Number-conserving strings of creation and annihilation operators

A physical electronic Hamiltonian preserves electron number:

$$
[\hat{\mathcal N},\hat H]=0.
$$

The simplest product of ladder operators that preserves particle number is

$$
a_p^\dagger a_q.
$$

Using

$$
[\hat{\mathcal N},AB]
=
[\hat{\mathcal N},A]B
+
A[\hat{\mathcal N},B],
$$

we obtain

$$
[\hat{\mathcal N},a_p^\dagger a_q]
=
a_p^\dagger a_q
-
a_p^\dagger a_q
=
0.
$$

Its physical action is transparent:

$$
\boxed{
a_p^\dagger a_q:
\quad
\text{remove an electron from }q
\text{ and create it in }p.
}
$$

This is a **one-orbital replacement operator**.

Similarly,

$$
a_p^\dagger a_q^\dagger a_s a_r
$$

first removes electrons from \(r\) and \(s\), then creates electrons in \(p\) and \(q\):

$$
\boxed{
a_p^\dagger a_q^\dagger a_s a_r:
\quad
(r,s)\longrightarrow(p,q).
}
$$

It is therefore the elementary **two-orbital replacement operator**.

More generally,

$$
a_{p_1}^\dagger \cdots a_{p_k}^\dagger
a_{q_k}\cdots a_{q_1}
$$

preserves particle number and changes at most \(k\) occupied modes. It is the elementary operator structure of a \(k\)-body contribution.

# Part I. One-body operators

## 3. A one-particle operator in first quantization

Begin in the one-particle Hilbert space. Let \(\hat o\) be any one-particle operator and let the orthonormal spin-orbitals satisfy

$$
\langle \chi_p|\chi_q\rangle=\delta_{pq}.
$$

Insert the resolution of the identity,

$$
\hat 1_1
=
\sum_p |\chi_p\rangle\langle\chi_p|.
$$

The operator can be expanded as

$$
\hat o
=
\hat 1_1\,\hat o\,\hat 1_1
=
\sum_{pq}
|\chi_p\rangle
\langle\chi_p|\hat o|\chi_q\rangle
\langle\chi_q|.
$$

Define the one-particle matrix elements

$$
\boxed{
o_{pq}
=
\langle\chi_p|\hat o|\chi_q\rangle.
}
$$

Then

$$
\boxed{
\hat o
=
\sum_{pq}
o_{pq}
|\chi_p\rangle\langle\chi_q|.
}
$$

The dyadic operator

$$
|\chi_p\rangle\langle\chi_q|
$$

takes a particle in mode \(q\) and places it in mode \(p\). This is exactly the role played by \(a_p^\dagger a_q\) in Fock space.

## 4. From \(|p\rangle\langle q|\) to \(a_p^\dagger a_q\)

Define the one-particle Fock-space state

$$
|q\rangle
=
a_q^\dagger|0\rangle.
$$

Now act with \(a_p^\dagger a_q\) on a one-particle state \(|r\rangle\):

$$
a_p^\dagger a_q |r\rangle
=
a_p^\dagger a_q a_r^\dagger |0\rangle.
$$

Using

$$
a_q a_r^\dagger
=
\delta_{qr}
-
a_r^\dagger a_q,
$$

and

$$
a_q|0\rangle=0,
$$

we get

$$
a_p^\dagger a_q |r\rangle
=
\delta_{qr}a_p^\dagger|0\rangle
=
\delta_{qr}|p\rangle.
$$

But this is exactly the action of the first-quantized transition operator:

$$
|p\rangle\langle q|r\rangle
=
\delta_{qr}|p\rangle.
$$

Therefore,

$$
\boxed{
|p\rangle\langle q|
\quad\longrightarrow\quad
a_p^\dagger a_q.
}
$$

This is the fundamental replacement rule for one-body operators.

## 5. General second-quantized one-body operator

The Fock-space representation of the one-particle operator \(\hat o\) is therefore

$$
\boxed{
\hat O^{(1)}
=
\sum_{pq}
o_{pq}
a_p^\dagger a_q.
}
$$

Let us verify its one-particle matrix elements:

$$
\langle p|\hat O^{(1)}|q\rangle
=
\sum_{rs}
o_{rs}
\langle 0|
a_p a_r^\dagger a_s a_q^\dagger
|0\rangle.
$$

Since

$$
a_s a_q^\dagger|0\rangle=\delta_{sq}|0\rangle
$$

and

$$
\langle 0|a_p a_r^\dagger=\delta_{pr}\langle 0|,
$$

we obtain

$$
\langle p|\hat O^{(1)}|q\rangle
=
o_{pq}.
$$

Thus,

$$
\boxed{
\langle p|\hat O^{(1)}|q\rangle
=
\langle\chi_p|\hat o|\chi_q\rangle.
}
$$

Second quantization reproduces the original one-particle matrix exactly, while extending the operator automatically to all particle-number sectors.

## 6. How the one-body operator acts on many-electron determinants

Let

$$
|\Phi\rangle
=
a_{i_1}^\dagger a_{i_2}^\dagger \cdots a_{i_{N_e}}^\dagger|0\rangle
$$

be a determinant, with occupied spin-orbitals \(i_1,\ldots,i_{N_e}\).

In

$$
\hat O^{(1)}
=
\sum_{pq}o_{pq}a_p^\dagger a_q,
$$

the annihilation operator \(a_q\) gives zero unless \(q\) is occupied. When \(q\) is occupied, it removes this occupation. Then \(a_p^\dagger\) creates occupation in \(p\), unless \(p\) is already occupied.

Therefore a one-body operator produces:

1. diagonal contributions, \(p=q\), which leave the occupation pattern unchanged;
2. singly replaced determinants, \(q\to p\), which differ from the original determinant in one occupied spin-orbital.

It cannot directly generate a determinant differing in two or more occupations.

$$
\boxed{
\text{A one-body operator connects determinants differing by at most one orbital occupation.}
}
$$

The fermionic sign associated with a replacement is automatically generated when the operators are anticommuted through the ordered string of occupied creation operators.

## 7. Concrete one-body example: a two-site hopping Hamiltonian

Consider two orthonormal modes \(|1\rangle\) and \(|2\rangle\), with one-particle Hamiltonian matrix

$$
\mathbf h
=
\begin{pmatrix}
\epsilon_1 & t \\
t^\ast & \epsilon_2
\end{pmatrix}.
$$

In first quantization,

$$
\hat h
=
\epsilon_1|1\rangle\langle 1|
+
\epsilon_2|2\rangle\langle 2|
+
t|1\rangle\langle 2|
+
t^\ast|2\rangle\langle 1|.
$$

In second quantization,

$$
\boxed{
\hat H_1
=
\epsilon_1 a_1^\dagger a_1
+
\epsilon_2 a_2^\dagger a_2
+
t a_1^\dagger a_2
+
t^\ast a_2^\dagger a_1.
}
$$

Using \(\hat n_i=a_i^\dagger a_i\),

$$
\hat H_1
=
\epsilon_1\hat n_1
+
\epsilon_2\hat n_2
+
t a_1^\dagger a_2
+
t^\ast a_2^\dagger a_1.
$$

The diagonal terms measure occupation energies. The off-diagonal terms move an electron between modes:

$$
a_1^\dagger a_2:
\quad
2\longrightarrow 1,
$$

$$
a_2^\dagger a_1:
\quad
1\longrightarrow 2.
$$

Thus a familiar one-particle matrix becomes an operator acting on the full Fock space.

# Part II. Two-body operators

## 8. A two-particle interaction in first quantization

Let \(\hat g(1,2)\) be an operator acting on two particles. For electron-electron Coulomb repulsion,

$$
\hat g(1,2)
=
\frac{1}{r_{12}}.
$$

In an \(N_e\)-electron system, every unordered electron pair contributes once:

$$
\boxed{
\hat W^{(2)}_{N_e}
=
\sum_{\alpha<\beta}
\hat g(\alpha,\beta).
}
$$

Equivalently,

$$
\hat W^{(2)}_{N_e}
=
\frac{1}{2}
\sum_{\alpha\neq\beta}
\hat g(\alpha,\beta).
$$

In an orthonormal spin-orbital basis, define the ordinary ordered two-particle matrix element

$$
\boxed{
(pq|rs)
=
\iint dx_1\,dx_2\,
\chi_p^\ast(x_1)\chi_q^\ast(x_2)
\,\hat g(1,2)\,
\chi_r(x_1)\chi_s(x_2).
}
$$

This is the amplitude for the two-particle replacement

$$
(r,s)\longrightarrow(p,q)
$$

before the fermionic exchange symmetry is accounted for in the two-particle state.

## 9. Pair replacements in Fock space

The Fock-space operator that removes electrons from modes \(r\) and \(s\) and creates electrons in modes \(p\) and \(q\) is

$$
a_p^\dagger a_q^\dagger a_s a_r.
$$

The annihilation order is reversed relative to the initial creation order so that, for distinct \(r,s\),

$$
a_s a_r
\left(
a_r^\dagger a_s^\dagger|0\rangle
\right)
=
|0\rangle.
$$

The second-quantized representation of a symmetric pair interaction is

$$
\boxed{
\hat W^{(2)}
=
\frac{1}{2}
\sum_{pqrs}
(pq|rs)\,
a_p^\dagger a_q^\dagger a_s a_r.
}
$$

The factor \(1/2\) is inherited from the first-quantized sum over unordered particle pairs:

$$
\sum_{\alpha<\beta}
=
\frac{1}{2}\sum_{\alpha\neq\beta}.
$$

It ensures that the two-particle interaction is counted once rather than twice.

## 10. Matrix element between two-electron determinants

Define ordered two-fermion determinant states

$$
|rs\rangle
=
a_r^\dagger a_s^\dagger|0\rangle,
\qquad
\langle pq|
=
\langle0|a_q a_p.
$$

We evaluate

$$
\langle pq|\hat W^{(2)}|rs\rangle.
$$

Insert the operator expansion:

$$
\langle pq|\hat W^{(2)}|rs\rangle
=
\frac{1}{2}
\sum_{tuvw}
(tu|vw)
\langle0|
a_q a_p
a_t^\dagger a_u^\dagger
a_w a_v
a_r^\dagger a_s^\dagger
|0\rangle.
$$

The annihilation operators acting on the ket give

$$
a_w a_v a_r^\dagger a_s^\dagger|0\rangle
=
\left(
\delta_{vr}\delta_{ws}
-
\delta_{vs}\delta_{wr}
\right)|0\rangle.
$$

Similarly, the contraction with the bra gives

$$
\langle0|a_q a_p a_t^\dagger a_u^\dagger
=
\left(
\delta_{pt}\delta_{qu}
-
\delta_{pu}\delta_{qt}
\right)\langle0|.
$$

Therefore,

$$
\begin{aligned}
\langle pq|\hat W^{(2)}|rs\rangle
=
\frac{1}{2}\big[
&(pq|rs)
-
(pq|sr)\\
&-
(qp|rs)
+
(qp|sr)
\big].
\end{aligned}
$$

For a symmetric interaction such as the Coulomb interaction,

$$
(qp|sr)=(pq|rs),
\qquad
(qp|rs)=(pq|sr).
$$

Hence,

$$
\boxed{
\langle pq|\hat W^{(2)}|rs\rangle
=
(pq|rs)-(pq|sr).
}
$$

## 11. Antisymmetrized two-electron integrals: direct minus exchange

Define the antisymmetrized two-electron integral

$$
\boxed{
\langle pq||rs\rangle
=
(pq|rs)-(pq|sr).
}
$$

Then the two-electron determinant matrix element becomes

$$
\boxed{
\langle pq|\hat W^{(2)}|rs\rangle
=
\langle pq||rs\rangle.
}
$$

The two contributions have distinct physical meanings:

$$
(pq|rs)
\quad\text{is the direct contribution},
$$

$$
(pq|sr)
\quad\text{is the exchange contribution}.
$$

Most importantly, exchange is not an extra classical force added to Coulomb repulsion. It arises because electrons occupy antisymmetric many-particle states, equivalently because their creation and annihilation operators anticommute.

$$
\boxed{
\text{fermionic antisymmetry}
\quad\Longrightarrow\quad
\text{direct minus exchange}.
}
$$

## 12. The same direct-minus-exchange result in first quantization

The normalized antisymmetric two-electron state constructed from spin-orbitals \(r\) and \(s\) is

$$
\Phi_{rs}(1,2)
=
\frac{1}{\sqrt{2}}
\left[
\chi_r(1)\chi_s(2)
-
\chi_s(1)\chi_r(2)
\right].
$$

Similarly,

$$
\Phi_{pq}(1,2)
=
\frac{1}{\sqrt{2}}
\left[
\chi_p(1)\chi_q(2)
-
\chi_q(1)\chi_p(2)
\right].
$$

Evaluating the first-quantized matrix element gives

$$
\begin{aligned}
\langle \Phi_{pq}|\hat g(1,2)|\Phi_{rs}\rangle
=
\frac{1}{2}\big[
&(pq|rs)
-
(pq|sr)\\
&-
(qp|rs)
+
(qp|sr)
\big].
\end{aligned}
$$

For a symmetric interaction,

$$
\boxed{
\langle \Phi_{pq}|\hat g(1,2)|\Phi_{rs}\rangle
=
(pq|rs)-(pq|sr)
=
\langle pq||rs\rangle.
}
$$

Thus, first and second quantization give the same physics and the same matrix elements:

$$
\boxed{
\text{Second quantization is a change of representation, not a new theory.}
}
$$

## 13. Equivalent form using antisymmetrized integrals

Starting from

$$
\hat W^{(2)}
=
\frac{1}{2}
\sum_{pqrs}
(pq|rs)
a_p^\dagger a_q^\dagger a_s a_r,
$$

we may use the antisymmetry of the operator string,

$$
a_p^\dagger a_q^\dagger
=
-
a_q^\dagger a_p^\dagger,
$$

to collect direct and exchange contributions. The result is

$$
\boxed{
\hat W^{(2)}
=
\frac{1}{4}
\sum_{pqrs}
\langle pq||rs\rangle
a_p^\dagger a_q^\dagger a_s a_r.
}
$$

Both expressions are identical:

$$
\boxed{
\frac{1}{2}\sum_{pqrs}(pq|rs)
a_p^\dagger a_q^\dagger a_s a_r
=
\frac{1}{4}\sum_{pqrs}\langle pq||rs\rangle
a_p^\dagger a_q^\dagger a_s a_r.
}
$$

The prefactors must be paired with the chosen integral convention:

$$
\begin{array}{ccl}
\text{ordinary integrals }(pq|rs)
&\longrightarrow&
\frac{1}{2},\\[6pt]
\text{antisymmetrized integrals }\langle pq||rs\rangle
&\longrightarrow&
\frac{1}{4}.
\end{array}
$$

# Part III. General structure

## 14. General \(k\)-body operators

A \(k\)-body operator in first quantization has the structure

$$
\hat O^{(k)}_{N_e}
=
\sum_{\alpha_1<\cdots<\alpha_k}
\hat o^{(k)}(\alpha_1,\ldots,\alpha_k).
$$

Let

$$
o^{(k)}_{p_1\cdots p_k,\,q_1\cdots q_k}
=
\left\langle
\chi_{p_1}\cdots\chi_{p_k}
\middle|
\hat o^{(k)}
\middle|
\chi_{q_1}\cdots\chi_{q_k}
\right\rangle
$$

denote matrix elements in an ordered, un-antisymmetrized product basis.

Then its second-quantized representation is

$$
\boxed{
\hat O^{(k)}
=
\frac{1}{k!}
\sum_{\substack{p_1,\ldots,p_k\\q_1,\ldots,q_k}}
o^{(k)}_{p_1\cdots p_k,\,q_1\cdots q_k}
\,
a_{p_1}^\dagger\cdots a_{p_k}^\dagger
a_{q_k}\cdots a_{q_1}.
}
$$

The normal-ordered string

$$
a_{p_1}^\dagger\cdots a_{p_k}^\dagger
a_{q_k}\cdots a_{q_1}
$$

removes the occupied set \((q_1,\ldots,q_k)\) and replaces it by \((p_1,\ldots,p_k)\).

If fully antisymmetrized \(k\)-particle matrix elements are used instead, the prefactor convention changes. For \(k=2\), this is precisely the difference between the \(1/2\) ordinary-integral form and the \(1/4\) antisymmetrized-integral form.

## 15. Elegant continuous-space formulation: field operators

The replacement-operator derivation is sufficient and conceptually direct. A compact complementary formulation uses fermionic field operators.

Define

$$
\hat\psi(x)
=
\sum_p \chi_p(x)a_p,
\qquad
\hat\psi^\dagger(x)
=
\sum_p \chi_p^\ast(x)a_p^\dagger.
$$

The operator \(\hat\psi^\dagger(x)\) creates an electron at the continuous spin-space coordinate \(x\), while \(\hat\psi(x)\) annihilates one there.

A one-body operator is

$$
\hat O^{(1)}
=
\int dx\,
\hat\psi^\dagger(x)\,
\hat o(x)\,
\hat\psi(x).
$$

Inserting the orbital expansion gives

$$
\begin{aligned}
\hat O^{(1)}
&=
\sum_{pq}
\left[
\int dx\,
\chi_p^\ast(x)\hat o(x)\chi_q(x)
\right]
a_p^\dagger a_q\\
&=
\sum_{pq}
o_{pq}a_p^\dagger a_q.
\end{aligned}
$$

A two-body interaction is

$$
\hat W^{(2)}
=
\frac{1}{2}
\iint dx_1\,dx_2\,
\hat\psi^\dagger(x_1)
\hat\psi^\dagger(x_2)
\hat g(1,2)
\hat\psi(x_2)
\hat\psi(x_1).
$$

Inserting the orbital expansion gives

$$
\hat W^{(2)}
=
\frac{1}{2}
\sum_{pqrs}
(pq|rs)
a_p^\dagger a_q^\dagger a_s a_r.
$$

This formulation makes the bridge from continuous first quantization to orbital-based second quantization almost immediate.

## 16. Summary: the conceptual bridge

The logic of operator construction is:

$$
\boxed{
\begin{array}{c}
\text{First-quantized operator acting on one particle}\\[4pt]
\hat o,\quad o_{pq}=\langle\chi_p|\hat o|\chi_q\rangle\\[4pt]
\Downarrow\\[4pt]
\text{One-particle replacement on Fock space}\\[4pt]
\hat O^{(1)}=\displaystyle\sum_{pq}o_{pq}a_p^\dagger a_q
\end{array}
}
$$

and

$$
\boxed{
\begin{array}{c}
\text{First-quantized pair interaction}\\[4pt]
\hat g(1,2),\quad (pq|rs)=\langle pq|\hat g|rs\rangle\\[4pt]
\Downarrow\\[4pt]
\text{Two-particle replacement on Fock space}\\[4pt]
\hat W^{(2)}
=
\displaystyle\frac{1}{2}\sum_{pqrs}
(pq|rs)a_p^\dagger a_q^\dagger a_s a_r\\[4pt]
\Downarrow\\[4pt]
\text{Fermionic determinant matrix elements}\\[4pt]
\langle pq|\hat W^{(2)}|rs\rangle
=
(pq|rs)-(pq|sr)
=
\langle pq||rs\rangle
\end{array}
}
$$

The central lesson is:

> A one-body integral is the amplitude for replacing one occupied spin-orbital by another. A two-body integral is the amplitude for replacing an occupied spin-orbital pair. For electrons, antisymmetry converts the ordinary pair matrix elements into direct-minus-exchange matrix elements.
