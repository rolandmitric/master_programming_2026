# Introduction to Second Quantization

We consider a system with $L$ discrete sites. These sites could represent literally spatially localized sites, molecular orbitals, or some other kind of modes. We assume that each site can be either empty or occupied by particles. The simplest physical state of such a system is the state with no particles at all. This state is called the vacuum state and is denoted by

$$
|0\rangle .
$$

By adding a single particle to the vacuum, we obtain a state with one particle occupying site $i$. We therefore introduce an operator $a_i^\dagger$ whose action on the vacuum is

$$
a_i^\dagger |0\rangle = |i\rangle .
$$

Here $|i\rangle$ denotes the normalized one-particle state in which the particle occupies site $i$.

However, this equation alone is only a partial definition of $a_i^\dagger$. It only tells us what $a_i^\dagger$ does to the vacuum. If $a_i^\dagger$ is supposed to be a genuine creation operator, it must be defined by its action on the whole many-particle Hilbert space. In particular, we assume that $a_i^\dagger$ raises the particle number in mode $i$ by one.

That is, $a_i^\dagger$ maps an $N$-particle state to an $(N+1)$-particle state:

$$
a_i^\dagger : \mathcal{H}_N \to \mathcal{H}_{N+1}.
$$

The operator $a_i$ is defined as the adjoint of $a_i^\dagger$:

$$
a_i = (a_i^\dagger)^\dagger .
$$

Therefore $a_i$ has the opposite action on particle-number sectors. If $a_i^\dagger$ raises the particle number by one, then its adjoint $a_i$ lowers the particle number by one:

$$
a_i : \mathcal{H}_{N+1} \to \mathcal{H}_N .
$$

This is the basic reason why $a_i$ is called an annihilation operator.

We can now show this explicitly for the one-particle state $|i\rangle$. Starting from

$$
a_i^\dagger |0\rangle = |i\rangle ,
$$

we take the adjoint. This gives

$$
\langle 0|a_i = \langle i| .
$$

Now we multiply this equation from the right by $|i\rangle$:

$$
\langle 0|a_i|i\rangle = \langle i|i\rangle .
$$

Since $|i\rangle$ is normalized,

$$
\langle i|i\rangle = 1 .
$$

Therefore

$$
\langle 0|a_i|i\rangle = 1 .
$$

Because $a_i$ lowers the particle number by one, the state $a_i|i\rangle$ must lie in the zero-particle sector. But the zero-particle sector is one-dimensional and is spanned only by the vacuum state:

$$
\mathcal{H}_0 = \operatorname{span}\{|0\rangle\}.
$$

Hence $a_i|i\rangle$ must be proportional to the vacuum:

$$
a_i|i\rangle = c |0\rangle .
$$

To determine the constant $c$, we take the scalar product with $\langle 0|$:

$$
\langle 0|a_i|i\rangle = c \langle 0|0\rangle .
$$

Since the vacuum is normalized,

$$
\langle 0|0\rangle = 1 ,
$$

we obtain

$$
\langle 0|a_i|i\rangle = c .
$$

But we already found that

$$
\langle 0|a_i|i\rangle = 1 .
$$

Therefore

$$
c=1 .
$$

Thus

$$
a_i|i\rangle = |0\rangle .
$$

This shows that $a_i$ removes one particle from site $i$. In other words, $a_i$ annihilates the particle created by $a_i^\dagger$.

If the particle is in a different site $j\neq i$, then there is no particle in mode $i$ that can be removed. Therefore

$$
a_i|j\rangle = 0
\qquad \text{for } i\neq j .
$$

Together, the action on one-particle states is therefore

$$
a_i|j\rangle = \delta_{ij}|0\rangle .
$$

At this stage, we have not yet decided whether the particles are bosons or fermions. The above reasoning only uses the fact that $a_i^\dagger$ raises the occupation of mode $i$ by one and that $a_i$ is its adjoint. The distinction between bosons and fermions enters later, when we ask how different creation operators behave relative to each other.

For bosons, different creation and annihilation operators satisfy commutation relations. For fermions, they satisfy anticommutation relations. But the basic one-particle interpretation is the same:

$$
a_i^\dagger |0\rangle = |i\rangle ,
$$

and

$$
a_i |i\rangle = |0\rangle .
$$

Thus $a_i^\dagger$ creates a particle in mode $i$, while $a_i$ annihilates a particle in mode $i$.


## 1. Particle-number sectors and Fock space

Starting from the vacuum, successive applications of creation operators generate states containing different numbers of particles:

$$
|0\rangle ,
$$

$$
a_i^\dagger |0\rangle ,
$$

$$
a_i^\dagger a_j^\dagger |0\rangle ,
$$

$$
a_i^\dagger a_j^\dagger a_k^\dagger |0\rangle ,
$$

and so forth.

These states naturally belong to different subspaces:

$$
\mathcal H_0 = \text{zero-particle sector},
$$

$$
\mathcal H_1 = \text{one-particle sector},
$$

$$
\mathcal H_2 = \text{two-particle sector},
$$

$$
\ldots
$$

The full state space that accommodates arbitrary particle numbers is the **Fock space**:

$$
\mathcal F
=
\mathcal H_0
\oplus
\mathcal H_1
\oplus
\mathcal H_2
\oplus \cdots .
$$

A state \(|\Psi_n\rangle \in \mathcal H_n\) contains precisely \(n\) particles.



## 2. The total number operator

Particle number is a measurable quantity. It is therefore represented by an operator, denoted by

$$
\hat{\mathcal N}.
$$

Its defining property is that every state in the \(n\)-particle sector is an eigenstate with eigenvalue \(n\):

$$
\boxed{
\hat{\mathcal N}|\Psi_n\rangle
=
n|\Psi_n\rangle,
\qquad
|\Psi_n\rangle \in \mathcal H_n.
}
$$

In particular,

$$
\hat{\mathcal N}|0\rangle = 0,
$$

$$
\hat{\mathcal N}a_i^\dagger|0\rangle
=
a_i^\dagger|0\rangle,
$$

and

$$
\hat{\mathcal N}a_i^\dagger a_j^\dagger|0\rangle
=
2a_i^\dagger a_j^\dagger|0\rangle.
$$

If \(P_n\) denotes the projector onto the \(n\)-particle subspace \(\mathcal H_n\), then the total number operator can be formally written as

$$
\boxed{
\hat{\mathcal N}
=
\sum_{n=0}^{\infty} n P_n.
}
$$

This is the fundamental definition of the number operator. It is independent of whether the particles are bosons or fermions.



## 3. Creation means raising the particle number by one

The physical meaning of a creation operator is that it maps an \(n\)-particle state to an \((n+1)\)-particle state:

$$
\boxed{
a_i^\dagger : \mathcal H_n \longrightarrow \mathcal H_{n+1}.
}
$$

Let

$$
|\Psi_n\rangle \in \mathcal H_n.
$$

Then

$$
\hat{\mathcal N}|\Psi_n\rangle
=
n|\Psi_n\rangle.
$$

Since \(a_i^\dagger|\Psi_n\rangle\) contains one additional particle,

$$
\hat{\mathcal N}a_i^\dagger|\Psi_n\rangle
=
(n+1)a_i^\dagger|\Psi_n\rangle.
$$

On the other hand,

$$
a_i^\dagger\hat{\mathcal N}|\Psi_n\rangle
=
n a_i^\dagger|\Psi_n\rangle.
$$

Subtracting the two expressions gives

$$
\left(
\hat{\mathcal N}a_i^\dagger
-
a_i^\dagger\hat{\mathcal N}
\right)|\Psi_n\rangle
=
a_i^\dagger|\Psi_n\rangle.
$$

Since this relation holds for every particle-number eigenstate, it becomes the operator identity

$$
\boxed{
[\hat{\mathcal N},a_i^\dagger]
=
a_i^\dagger.
}
$$

This commutator is the precise algebraic formulation of the statement:

$$
\boxed{
a_i^\dagger \text{ creates one particle.}
}
$$



## 4. Annihilation means lowering the particle number by one

Analogously, an annihilation operator \(a_i\) removes one particle:

$$
\boxed{
a_i : \mathcal H_n \longrightarrow \mathcal H_{n-1}.
}
$$

For an \(n\)-particle state,

$$
\hat{\mathcal N}a_i|\Psi_n\rangle
=
(n-1)a_i|\Psi_n\rangle,
$$

while

$$
a_i\hat{\mathcal N}|\Psi_n\rangle
=
n a_i|\Psi_n\rangle.
$$

Therefore,

$$
\boxed{
[\hat{\mathcal N},a_i]
=
-a_i.
}
$$

Thus the number operator distinguishes raising and lowering operations:

$$
[\hat{\mathcal N},a_i^\dagger]=a_i^\dagger,
\qquad
[\hat{\mathcal N},a_i]=-a_i.
$$



## 5. Local occupation numbers

The total number operator counts all particles in the system. Since the system consists of discrete sites, we may also ask how many particles occupy a particular site \(i\).

Introduce occupation-number basis states

$$
|n_1,n_2,\ldots,n_L\rangle ,
$$

where \(n_i\) is the occupation of site \(i\).

The **local number operator** \(\hat n_i\) is defined by

$$
\boxed{
\hat n_i
|n_1,\ldots,n_i,\ldots,n_L\rangle
=
n_i
|n_1,\ldots,n_i,\ldots,n_L\rangle.
}
$$

The total particle number is the sum of all local occupations:

$$
\boxed{
\hat{\mathcal N}
=
\sum_{i=1}^{L}\hat n_i.
}
$$

For example, the state

$$
|0,1,0,2\rangle
$$

contains one particle on site \(2\) and two particles on site \(4\). Hence

$$
\hat n_2|0,1,0,2\rangle
=
|0,1,0,2\rangle,
$$

$$
\hat n_4|0,1,0,2\rangle
=
2|0,1,0,2\rangle,
$$

and

$$
\hat{\mathcal N}|0,1,0,2\rangle
=
3|0,1,0,2\rangle.
$$



## 6. Creation at a particular site

The operator \(a_j^\dagger\) does not merely add one particle somewhere. It adds one particle specifically at site \(j\).

Therefore, it increases the occupation of site \(j\) by one, while leaving all other local occupations unchanged:

$$
a_j^\dagger :
|n_1,\ldots,n_j,\ldots,n_L\rangle
\longrightarrow
|n_1,\ldots,n_j+1,\ldots,n_L\rangle .
$$

This implies

$$
\boxed{
[\hat n_i,a_j^\dagger]
=
\delta_{ij}a_j^\dagger.
}
$$

For \(i=j\),

$$
[\hat n_j,a_j^\dagger]
=
a_j^\dagger,
$$

because creation at \(j\) raises the local occupation by one.

For \(i\neq j\),

$$
[\hat n_i,a_j^\dagger]
=
0,
$$

because creation at \(j\) does not change the occupation at any other site.

Similarly,

$$
\boxed{
[\hat n_i,a_j]
=
-\delta_{ij}a_j.
}
$$



## 7. Obtaining \(\hat n_i=a_i^\dagger a_i\)

So far, the number operator has been introduced independently as the observable that counts particles. We have not yet identified it with \(a_i^\dagger a_i\).

To do this, one must specify the normalization of the ladder action.

### Bosonic occupation numbers

For bosons, any non-negative occupation number is allowed:

$$
n_i=0,1,2,\ldots
$$

The conventional normalized ladder action is

$$
a_i^\dagger
|n_1,\ldots,n_i,\ldots,n_L\rangle
=
\sqrt{n_i+1}
|n_1,\ldots,n_i+1,\ldots,n_L\rangle,
$$

and

$$
a_i
|n_1,\ldots,n_i,\ldots,n_L\rangle
=
\sqrt{n_i}
|n_1,\ldots,n_i-1,\ldots,n_L\rangle.
$$

Applying first \(a_i\) and then \(a_i^\dagger\) gives

$$
a_i^\dagger a_i
|n_1,\ldots,n_i,\ldots,n_L\rangle
=
n_i
|n_1,\ldots,n_i,\ldots,n_L\rangle.
$$

But this is exactly the defining action of \(\hat n_i\). Therefore,

$$
\boxed{
\hat n_i=a_i^\dagger a_i.
}
$$

Consequently,

$$
\boxed{
\hat{\mathcal N}
=
\sum_{i=1}^{L}a_i^\dagger a_i.
}
$$



### Fermionic occupation numbers

For fermions, the local occupations are restricted to

$$
n_i\in\{0,1\}.
$$

The Pauli principle forbids creation on an already occupied mode:

$$
a_i^\dagger |n_i=1\rangle = 0.
$$

Nevertheless, on the allowed occupation-number states, the product \(a_i^\dagger a_i\) again measures whether site \(i\) is occupied:

$$
a_i^\dagger a_i |n_i=0\rangle = 0,
$$

$$
a_i^\dagger a_i |n_i=1\rangle = |n_i=1\rangle.
$$

Thus also for fermions,

$$
\boxed{
\hat n_i=a_i^\dagger a_i,
\qquad
\hat{\mathcal N}=\sum_i a_i^\dagger a_i.
}
$$



## 8. What the number operator does not yet determine

The number operator makes precise what is meant by creating or annihilating particles:

$$
[\hat{\mathcal N},a_i^\dagger]=a_i^\dagger,
\qquad
[\hat{\mathcal N},a_i]=-a_i.
$$

However, these relations do **not** by themselves determine how two creation operators behave when their order is interchanged:

$$
a_i^\dagger a_j^\dagger
\quad \text{versus} \quad
a_j^\dagger a_i^\dagger.
$$

That question concerns the exchange properties of identical particles:

- For bosons, many-particle states are symmetric under exchange and the creation operators commute.
- For fermions, many-particle states are antisymmetric under exchange and the creation operators anticommute.

Thus the logical order is:

$$
\boxed{
\text{Fock-space sectors}
\;\longrightarrow\;
\text{number operator}
\;\longrightarrow\;
\text{raising/lowering relations}
\;\longrightarrow\;
\text{bosonic or fermionic exchange algebra}.
}
$$



## Summary

The number operator should be motivated as the physical observable that counts particles in Fock space:

$$
\hat{\mathcal N}|\Psi_n\rangle=n|\Psi_n\rangle.
$$

The defining meaning of creation and annihilation operators is then expressed through

$$
\boxed{
[\hat{\mathcal N},a_i^\dagger]=a_i^\dagger,
\qquad
[\hat{\mathcal N},a_i]=-a_i.
}
$$

For discrete sites, local occupations are described by

$$
\hat n_i
|n_1,\ldots,n_i,\ldots,n_L\rangle
=
n_i
|n_1,\ldots,n_i,\ldots,n_L\rangle,
$$

with

$$
\hat{\mathcal N}=\sum_i \hat n_i.
$$

After choosing the normalized ladder action, one obtains

$$
\boxed{
\hat n_i=a_i^\dagger a_i,
\qquad
\hat{\mathcal N}=\sum_i a_i^\dagger a_i.
}
$$

The conceptual point is that the number operator is not introduced merely because the expression \(a_i^\dagger a_i\) can be written. It is introduced because it provides the observable whose eigenvalues make the statement “\(a_i^\dagger\) adds one particle” mathematically precise.
