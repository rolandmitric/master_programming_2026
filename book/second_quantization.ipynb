# Introduction to Second Quantization

We consider a system tha consist of some number ($L$) of discrete sites. These sites could represent literally spatially localized sites than can host particles, molecular orbitals, or some other kind of general states (modes). 

We assume that each of such sites can be either empty or occupied by some number particles.
The simplest physical state of such a system is the state with no particles at all. This state is called the vacuum state and is often denoted by

$$
|0\rangle .
$$

By adding a single particle to the vacuum, we obtain a state with one particle occupying one of the possible sites $i$. We formally introduce a linear (hermitian) operator $a_i^\dagger$ whose action on the vacuum is defined by

$$
a_i^\dagger |0\rangle = |i\rangle .
$$



Thus the operator $a_i^\dagger$ creates a particle in mode $i$ and fills the state $|i\rangle$.

However, this equation alone is only a partial definition of $a_i^\dagger$. It only tells us what $a_i^\dagger$ does to the vacuum state. If $a_i^\dagger$ is supposed to be a genuine creation operator, it must be defined by its action on the whole many-particle Hilbert space
which will be the case if its action on the basis states is specified. We assume that the many-particle Hilbert space is a direct sum of subspaces with different particle numbers:

$$
\mathcal{H} = \bigoplus_{N=0}^\infty \mathcal{H}_N ,
$$

where $\mathcal{H}_N$ is the $N$-particle sector of the Hilbert space. The vacuum state $|0\rangle$ is the only state in the zero-particle sector $\mathcal{H}_0$. The one-particle sector $\mathcal{H}_1$ is spanned by the states $|i\rangle$ for all sites $i$. The two-particle sector $\mathcal{H}_2$ is spanned by states with two particles, and so on. Such space has been introduced into physics by Dirac and Jordan and Fock in the 1920s and is called the Fock space. The Fock space is a convenient way to describe systems with variable particle numbers.

Having defined the Fock space, we can now define the action of $a_i^\dagger$ on the whole space. In particular, we assume that $a_i^\dagger$ raises the particle number in mode $i$ by one regardless of the particle number in other modes. That is, $a_i^\dagger$ maps an $N$-particle state to an $(N+1)$-particle state:

$$
a_i^\dagger : \mathcal{H}_N \to \mathcal{H}_{N+1}.
$$

$$
a_i^\dagger |n_1, n_2, \ldots, n_i, \ldots, n_L\rangle \propto |n_1, n_2, \ldots, n_i + 1, \ldots,n_L\rangle .
$$



The fact that we have defined $a_i^\dagger$ already suggests that there should be an operator $a_i$ that does the opposite. The operator $a_i$ is defined as the adjoint of $a_i^\dagger$:

$$
a_i = (a_i^\dagger)^\dagger .
$$

If $a_i^\dagger$ raises the particle number by one, then its adjoint $a_i$ lowers the particle number by one:

$$
a_i : \mathcal{H}_{N+1} \to \mathcal{H}_N .
$$

Thus, $a_i$ is called an annihilation operator and it removes a particle from mode $i$.
We can now show this explicitly for the one-particle state $|i\rangle$. Starting from

$$
a_i^\dagger |0\rangle = |i\rangle ,
$$

we take the adjoint. This gives

$$
\langle 0|a_i = \langle i| .
$$

Now we multiply this equation from the right by $|i\rangle$:

$$
\langle 0|a_i|i\rangle = \langle i|i\rangle .
$$

Since $|i\rangle$ is normalized,

$$
\langle i|i\rangle = 1 .
$$

Therefore

$$
\langle 0|a_i|i\rangle = 1 .
$$

Because $a_i$ lowers the particle number by one, the state $a_i|i\rangle$ must lie in the zero-particle sector. But the zero-particle sector is one-dimensional and is spanned only by the vacuum state:

$$
\mathcal{H}_0 = \operatorname{span}\{|0\rangle\}.
$$

Hence $a_i|i\rangle$ must be proportional to the vacuum:

$$
a_i|i\rangle = c |0\rangle .
$$

To determine the constant $c$, we take the scalar product with $\langle 0|$:

$$
\langle 0|a_i|i\rangle = c \langle 0|0\rangle .
$$

Since the vacuum is normalized,

$$
\langle 0|0\rangle = 1 ,
$$

we obtain

$$
\langle 0|a_i|i\rangle = c .
$$

But we already found that

$$
\langle 0|a_i|i\rangle = 1 .
$$

Therefore

$$
c=1 .
$$

Thus

$$
a_i|i\rangle = |0\rangle .
$$

This shows that $a_i$ removes one particle from site $i$. In other words, $a_i$ annihilates the particle created by $a_i^\dagger$.

If the particle is in a different site $j\neq i$, then there is no particle in mode $i$ that can be removed. Therefore

$$
a_i|j\rangle = 0
\qquad \text{for } i\neq j .
$$

Together, the action on one-particle states is therefore

$$
a_i|j\rangle = \delta_{ij}|0\rangle .
$$

At this stage, we have not yet decided whether the particles are bosons or fermions. The above reasoning only uses the fact that $a_i^\dagger$ raises the occupation of mode $i$ by one and that $a_i$ is its adjoint. 

## From number operators to bosons and fermions

We have introduced occupation-number states

$$
|n_1,n_2,\ldots,n_L\rangle
$$

that contain a definite number of particles in each mode. Here $L$ is the number of available modes, and $n_i$ denotes the occupation number of mode $i$. We now want to construct an operator $\hat n_i$ that counts the number of particles in mode $i$. Therefore we require

$$
\hat n_i |n_1,\ldots,n_i,\ldots,n_L\rangle
=
n_i |n_1,\ldots,n_i,\ldots,n_L\rangle .
$$

The eigenvalue $n_i$ is the occupation number of mode $i$. For bosons, $n_i$ can be any non-negative integer,

$$
n_i = 0,1,2,\ldots ,
$$

whereas for fermions, the Pauli principle restricts the allowed occupation numbers to

$$
n_i = 0,1 .
$$

A creation operator for mode $i$ should raise the occupation number of mode $i$ by one,

$$
a_i^\dagger |n_1,\ldots,n_i,\ldots,n_L\rangle
\propto
|n_1,\ldots,n_i+1,\ldots,n_L\rangle ,
$$

whereas an annihilation operator should lower the occupation number by one,

$$
a_i |n_1,\ldots,n_i,\ldots,n_L\rangle
\propto
|n_1,\ldots,n_i-1,\ldots,n_L\rangle .
$$

The proportionality constants are important. They determine the normalization of the many-particle states and the spectrum of the number operator.

The natural candidate for the number operator is

$$
\hat n_i = a_i^\dagger a_i .
$$


The operator $a_i$ first tries to remove one particle from mode $i$. If no particle is present, the result is zero. If particles are present, the state is lowered to one with one fewer particle in mode $i$. The operator $a_i^\dagger$ then puts the particle back into the same mode. Therefore the product $a_i^\dagger a_i$ returns the original state, multiplied by a factor that measures how many particles were initially present in mode $i$.



To see this explicitly, consider first a single bosonic mode. We write the occupation-number states as

$$
|0\rangle, |1\rangle, |2\rangle, \ldots .
$$

Assume that the annihilation operator acts as

$$
a|n\rangle = c_n |n-1\rangle ,
$$

where $$c_n$$ is a coefficient still to be determined. Since $$a^\dagger$$ is the Hermitian adjoint of $$a$$, we have correspondingly

$$
a^\dagger |n-1\rangle = c_n^* |n\rangle .
$$

Therefore

$$
a^\dagger a |n\rangle
=
a^\dagger c_n |n-1\rangle
=
|c_n|^2 |n\rangle .
$$

If $$a^\dagger a$$ is supposed to count particles, then we require

$$
a^\dagger a |n\rangle = n |n\rangle .
$$

Hence

$$
|c_n|^2 = n .
$$

Choosing the phases of the states conveniently, we take

$$
c_n = \sqrt n .
$$

Thus the bosonic annihilation operator is defined by

$$
a|n\rangle = \sqrt n\, |n-1\rangle ,
$$

and the corresponding creation operator acts as

$$
a^\dagger |n\rangle = \sqrt{n+1}\, |n+1\rangle .
$$

With these definitions,

$$
a^\dagger a |n\rangle
=
a^\dagger \sqrt n |n-1\rangle
=
\sqrt n \sqrt n |n\rangle
=
n |n\rangle .
$$

Therefore

$$
\hat n = a^\dagger a
$$

is indeed the number operator.

For many bosonic modes, the number operator for mode $i$ is

$$
\hat n_i = a_i^\dagger a_i .
$$

The annihilation operator acts as

$$
a_i |n_1,\ldots,n_i,\ldots,n_L\rangle
=
\sqrt{n_i}\,
|n_1,\ldots,n_i-1,\ldots,n_L\rangle ,
$$

and the creation operator acts as

$$
a_i^\dagger |n_1,\ldots,n_i,\ldots,n_L\rangle
=
\sqrt{n_i+1}\,
|n_1,\ldots,n_i+1,\ldots,n_L\rangle .
$$

Therefore

$$
a_i^\dagger a_i
|n_1,\ldots,n_i,\ldots,n_L\rangle
=
n_i
|n_1,\ldots,n_i,\ldots,n_L\rangle .
$$

This proves that $a_i^\dagger a_i$ counts the number of particles in mode $i$.


## Bosonic commutation relations

Let us now swap the order of the operators defining the number operator. The operator

$$
a_i a_i^\dagger
$$

does not count particles! For bosons,

$$
a_i a_i^\dagger |n_i\rangle
=
(n_i+1)|n_i\rangle .
$$

This happens because $$a_i^\dagger$$ first creates one additional particle and $$a_i$$ then removes it again. The result therefore contains the factor $$n_i+1$$, not $$n_i$$. By contrast, $$a_i^\dagger a_i$$ first removes a particle and then restores it. It therefore measures how many particles were present before the operation.

From this, we can derive the commutation relation for bosonic operators. We have

$$
[a_i, a_i^\dagger] = a_i a_i^\dagger - a_i^\dagger a_i = 1 .
$$

This is the fundamental commutation relation for bosons. For different modes $i \neq j$, the operators commute:

$$
[a_i, a_j] = [a_i^\dagger, a_j^\dagger] = [a_i, a_j^\dagger] = 0 .
$$` 

The commutation relations for bosonic operators are therefore

$$
[a_i, a_j] = [a_i^\dagger, a_j^\dagger] = 0, \quad [a_i, a_j^\dagger] = \delta_{ij} .
$$


## Fermionic anticommutation relations

For fermions, the logic is similar, but the allowed occupation numbers are restricted to

$$
n_i=0,1 .
$$

For a single fermionic mode,

$$
a|0\rangle = 0,
$$

because there is no particle to annihilate, and

$$
a|1\rangle = |0\rangle .
$$

Similarly,

$$
a^\dagger |0\rangle = |1\rangle ,
$$

but

$$
a^\dagger |1\rangle = 0,
$$

because a fermionic mode cannot be occupied twice.

Now apply $$a^\dagger a$$ to the two possible states. For the empty state,

$$
a^\dagger a |0\rangle
=
a^\dagger 0
=
0 .
$$

For the occupied state,

$$
a^\dagger a |1\rangle
=
a^\dagger |0\rangle
=
|1\rangle .
$$

Therefore

$$
a^\dagger a |0\rangle = 0 |0\rangle ,
$$

and

$$
a^\dagger a |1\rangle = 1 |1\rangle .
$$

Thus $$a^\dagger a$$ is also the fermionic number operator.




In the full fermionic occupation-number basis, one must additionally keep track of signs. We first choose an ordering of the modes, for example $$1,2,\ldots,L$$, and define

$$
|n_1,n_2,\ldots,n_L\rangle
=
(a_1^\dagger)^{n_1}
(a_2^\dagger)^{n_2}
\cdots
(a_L^\dagger)^{n_L}
|0\rangle .
$$

With this convention, the fermionic annihilation operator acts as

$$
a_i |n_1,\ldots,n_i,\ldots,n_L\rangle
=
\begin{cases}
0, & n_i=0,\\
(-1)^{\sum_{j<i} n_j}
|n_1,\ldots,n_i-1,\ldots,n_L\rangle, & n_i=1 .
\end{cases}
$$

The fermionic creation operator acts as

$$
a_i^\dagger |n_1,\ldots,n_i,\ldots,n_L\rangle
=
\begin{cases}
0, & n_i=1,\\
(-1)^{\sum_{j<i} n_j}
|n_1,\ldots,n_i+1,\ldots,n_L\rangle, & n_i=0 .
\end{cases}
$$

The sign factor

$$
(-1)^{\sum_{j<i} n_j}
$$

counts how many occupied modes lie before mode $$i$$. It appears because fermionic creation and annihilation operators anticommute.

However, in the product $$a_i^\dagger a_i$$, this sign appears twice and therefore cancels. Consequently,

$$
a_i^\dagger a_i
|n_1,\ldots,n_i,\ldots,n_L\rangle
=
n_i
|n_1,\ldots,n_i,\ldots,n_L\rangle .
$$

Thus, for both bosons and fermions, the number operator is

$$
\boxed{\hat n_i = a_i^\dagger a_i } .
$$

The difference between bosons and fermions is not the definition of the number operator. The difference lies in the allowed occupation numbers and in the algebra of the creation and annihilation operators.

For bosons,

$$
[a_i,a_j^\dagger]=\delta_{ij},
\qquad
[a_i,a_j]=[a_i^\dagger,a_j^\dagger]=0,
$$

and the occupation numbers are

$$
n_i=0,1,2,\ldots .
$$

For fermions,

$$
\{a_i,a_j^\dagger\}=\delta_{ij},
\qquad
\{a_i,a_j\}=\{a_i^\dagger,a_j^\dagger\}=0,
$$

and the occupation numbers are restricted to

$$
n_i=0,1 .
$$

Therefore $$a_i^\dagger a_i$$ is called the number operator because the occupation-number basis is its eigenbasis and the corresponding occupation numbers are its eigenvalues.

## Representing fermionic states with bitstrings

Since the occupation numbers of fermionic modes can only be $0$ or $1$, we can represent the occupation of a system of $L$ fermionic modes as a bitstring of length $L$. Each bit in the string corresponds to a mode, with `0` indicating that the mode is unoccupied and `1` indicating that it is occupied. This is convenient for both theoretical and computational purposes, as it allows us to efficiently encode the state of the system and perform operations on it. We now define a python class to represent fermionic states using bitstrings.

```python
class FermionFockSpace:
    def __init__(self, num_modes):
        self.num_modes = num_modes

    def vacuum(self):
        return 0

    def occupation(self, state, site):
        return (state >> site) & 1

    def count_particles(self, state):
        return state.bit_count()

    def parity_before(self, state, site):
        mask = (1 << site) - 1
        num_before = (state & mask).bit_count()
        return -1 if num_before % 2 else 1

    def create(self, state, site):
        if self.occupation(state, site) == 1:
            return 0, None

        coeff = self.parity_before(state, site)
        new_state = state | (1 << site)

        return coeff, new_state

    def annihilate(self, state, site):
        if self.occupation(state, site) == 0:
            return 0, None

        coeff = self.parity_before(state, site)
        new_state = state & ~(1 << site)

        return coeff, new_state

    def number(self, state, site):
        return self.occupation(state, site), state

    def state_to_occupations(self, state):
        return [(state >> i) & 1 for i in range(self.num_modes)]

    def occupations_to_state(self, occupations):
        state = 0
        for i, n_i in enumerate(occupations):
            if n_i:
                state |= (1 << i)
        return state

    def pretty(self, state):
        occs = self.state_to_occupations(state)
        return "|" + ",".join(str(n) for n in occs) + ">"
    
```

```python
fs = FermionFockSpace(num_modes=4)

state = fs.vacuum()
print(fs.pretty(state))

coeff, state = fs.create(state, 1)
print(coeff, fs.pretty(state))

coeff, state = fs.create(state, 3)
print(coeff, fs.pretty(state))

coeff, new_state = fs.create(state, 1)
print(coeff, new_state)

coeff, state = fs.annihilate(state, 1)
print(coeff, fs.pretty(state))
```
