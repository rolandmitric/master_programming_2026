# Introduction to Second Quantization

We consider a system with $L$ discrete sites. These sites could represent literally spatially localized sites, molecular orbitals, or some other kind of modes. We assume that each site can be either empty or occupied by particles. The simplest physical state of such a system is the state with no particles at all. This state is called the vacuum state and is denoted by

$$
|0\rangle .
$$

By adding a single particle to the vacuum, we obtain a state with one particle occupying site $i$. We therefore introduce an operator $a_i^\dagger$ whose action on the vacuum is

$$
a_i^\dagger |0\rangle = |i\rangle .
$$

Here $|i\rangle$ denotes the normalized one-particle state in which the particle occupies site $i$.

However, this equation alone is only a partial definition of $a_i^\dagger$. It only tells us what $a_i^\dagger$ does to the vacuum. If $a_i^\dagger$ is supposed to be a genuine creation operator, it must be defined by its action on the whole many-particle Hilbert space. In particular, we assume that $a_i^\dagger$ raises the particle number in mode $i$ by one.

That is, $a_i^\dagger$ maps an $N$-particle state to an $(N+1)$-particle state:

$$
a_i^\dagger : \mathcal{H}_N \to \mathcal{H}_{N+1}.
$$

The operator $a_i$ is defined as the adjoint of $a_i^\dagger$:

$$
a_i = (a_i^\dagger)^\dagger .
$$

Therefore $a_i$ has the opposite action on particle-number sectors. If $a_i^\dagger$ raises the particle number by one, then its adjoint $a_i$ lowers the particle number by one:

$$
a_i : \mathcal{H}_{N+1} \to \mathcal{H}_N .
$$

This is the basic reason why $a_i$ is called an annihilation operator.

We can now show this explicitly for the one-particle state $|i\rangle$. Starting from

$$
a_i^\dagger |0\rangle = |i\rangle ,
$$

we take the adjoint. This gives

$$
\langle 0|a_i = \langle i| .
$$

Now we multiply this equation from the right by $|i\rangle$:

$$
\langle 0|a_i|i\rangle = \langle i|i\rangle .
$$

Since $|i\rangle$ is normalized,

$$
\langle i|i\rangle = 1 .
$$

Therefore

$$
\langle 0|a_i|i\rangle = 1 .
$$

Because $a_i$ lowers the particle number by one, the state $a_i|i\rangle$ must lie in the zero-particle sector. But the zero-particle sector is one-dimensional and is spanned only by the vacuum state:

$$
\mathcal{H}_0 = \operatorname{span}\{|0\rangle\}.
$$

Hence $a_i|i\rangle$ must be proportional to the vacuum:

$$
a_i|i\rangle = c |0\rangle .
$$

To determine the constant $c$, we take the scalar product with $\langle 0|$:

$$
\langle 0|a_i|i\rangle = c \langle 0|0\rangle .
$$

Since the vacuum is normalized,

$$
\langle 0|0\rangle = 1 ,
$$

we obtain

$$
\langle 0|a_i|i\rangle = c .
$$

But we already found that

$$
\langle 0|a_i|i\rangle = 1 .
$$

Therefore

$$
c=1 .
$$

Thus

$$
a_i|i\rangle = |0\rangle .
$$

This shows that $a_i$ removes one particle from site $i$. In other words, $a_i$ annihilates the particle created by $a_i^\dagger$.

If the particle is in a different site $j\neq i$, then there is no particle in mode $i$ that can be removed. Therefore

$$
a_i|j\rangle = 0
\qquad \text{for } i\neq j .
$$

Together, the action on one-particle states is therefore

$$
a_i|j\rangle = \delta_{ij}|0\rangle .
$$

At this stage, we have not yet decided whether the particles are bosons or fermions. The above reasoning only uses the fact that $a_i^\dagger$ raises the occupation of mode $i$ by one and that $a_i$ is its adjoint. The distinction between bosons and fermions enters later, when we ask how different creation operators behave relative to each other.

For bosons, different creation and annihilation operators satisfy commutation relations. For fermions, they satisfy anticommutation relations. But the basic one-particle interpretation is the same:

$$
a_i^\dagger |0\rangle = |i\rangle ,
$$

and

$$
a_i |i\rangle = |0\rangle .
$$

Thus $a_i^\dagger$ creates a particle in mode $i$, while $a_i$ annihilates a particle in mode $i$.


## From number operators to bosons and fermions

We have introduced occupation-number states

$$
|n_1,n_2,\ldots,n_L\rangle
$$

and number operators $\hat n_i$ defined by

$$
\hat n_i |n_1,\ldots,n_i,\ldots,n_L\rangle
=
n_i |n_1,\ldots,n_i,\ldots,n_L\rangle .
$$

The eigenvalue $n_i$ is the number of particles occupying mode $i$. Therefore $n_i$ must be a non-negative integer.

We now ask what kind of creation and annihilation operators are compatible with this idea.

A creation operator for mode $i$ should raise the occupation number of mode $i$ by one:

$$
a_i^\dagger |n_1,\ldots,n_i,\ldots,n_L\rangle
\propto
|n_1,\ldots,n_i+1,\ldots,n_L\rangle .
$$

Similarly, an annihilation operator should lower the occupation number of mode $i$ by one:

$$
a_i |n_1,\ldots,n_i,\ldots,n_L\rangle
\propto
|n_1,\ldots,n_i-1,\ldots,n_L\rangle .
$$

Equivalently, the number operator should satisfy

$$
[\hat n_i,a_j^\dagger]
=
\delta_{ij}a_j^\dagger ,
$$

and

$$
[\hat n_i,a_j]
=
-\delta_{ij}a_j .
$$

The first relation says that $a_j^\dagger$ increases the occupation of mode $j$ by one. The second relation says that $a_j$ decreases the occupation of mode $j$ by one.

Now the question becomes: what algebra should the creation and annihilation operators satisfy?

There are two canonical possibilities.

### Bosonic case

For bosons, creation operators commute:

$$
[a_i^\dagger,a_j^\dagger]=0 .
$$

This means

$$
a_i^\dagger a_j^\dagger
=
a_j^\dagger a_i^\dagger .
$$

Thus creating one particle in mode $i$ and then one particle in mode $j$ gives the same state as creating them in the opposite order. The many-particle state is symmetric under exchange.

The full bosonic commutation relations are

$$
[a_i,a_j^\dagger]=\delta_{ij},
$$

and

$$
[a_i,a_j]=[a_i^\dagger,a_j^\dagger]=0.
$$

Define the number operator as

$$
\hat n_i = a_i^\dagger a_i .
$$

Using the commutator

$$
[a_i,a_i^\dagger]=1,
$$

one obtains

$$
[\hat n_i,a_i^\dagger]=a_i^\dagger .
$$

Indeed,

$$
[\hat n_i,a_i^\dagger]
=
[a_i^\dagger a_i,a_i^\dagger]
=
a_i^\dagger [a_i,a_i^\dagger]
=
a_i^\dagger .
$$

Therefore, if $|n_i\rangle$ is an eigenstate of $\hat n_i$ with eigenvalue $n_i$,

$$
\hat n_i |n_i\rangle = n_i |n_i\rangle ,
$$

then

$$
\hat n_i a_i^\dagger |n_i\rangle
=
(n_i+1) a_i^\dagger |n_i\rangle .
$$

Thus $a_i^\dagger |n_i\rangle$ is an eigenstate of $\hat n_i$ with eigenvalue $n_i+1$.

Similarly,

$$
[\hat n_i,a_i]=-a_i ,
$$

so $a_i$ lowers the occupation by one.

Starting from the vacuum state

$$
\hat n_i |0\rangle = 0,
$$

repeated application of $a_i^\dagger$ generates the ladder

$$
|0\rangle,\ |1\rangle,\ |2\rangle,\ |3\rangle,\ldots
$$

Thus the possible bosonic occupation numbers are

$$
n_i = 0,1,2,3,\ldots .
$$

For bosons, arbitrarily many particles can occupy the same mode.

The normalized bosonic ladder action is

$$
a_i^\dagger |n_i\rangle
=
\sqrt{n_i+1}|n_i+1\rangle ,
$$

and

$$
a_i |n_i\rangle
=
\sqrt{n_i}|n_i-1\rangle .
$$

Therefore

$$
a_i^\dagger a_i |n_i\rangle
=
n_i |n_i\rangle .
$$

Hence

$$
\hat n_i = a_i^\dagger a_i
$$

is indeed the operator that counts the number of bosons in mode $i$.

### Fermionic case

There is a second possibility. For fermions, creation operators anticommute:

$$
\{c_i^\dagger,c_j^\dagger\}=0 .
$$

This means

$$
c_i^\dagger c_j^\dagger
=
-
c_j^\dagger c_i^\dagger .
$$

Thus exchanging the order of two fermionic creation operators changes the sign of the state. The many-particle state is antisymmetric under exchange.

The full fermionic anticommutation relations are

$$
\{c_i,c_j^\dagger\}=\delta_{ij},
$$

and

$$
\{c_i,c_j\}=\{c_i^\dagger,c_j^\dagger\}=0.
$$

Now set $i=j$ in

$$
\{c_i^\dagger,c_i^\dagger\}=0 .
$$

This gives

$$
c_i^\dagger c_i^\dagger + c_i^\dagger c_i^\dagger = 0,
$$

or

$$
2(c_i^\dagger)^2=0 .
$$

Therefore

$$
(c_i^\dagger)^2=0 .
$$

This has an immediate physical consequence. If one tries to create two identical fermions in the same mode, one obtains zero:

$$
(c_i^\dagger)^2|0\rangle=0 .
$$

Thus a fermionic mode can be either empty or occupied by one particle, but not by two particles. The possible fermionic occupation numbers are therefore

$$
n_i=0,1 .
$$

Define the fermionic number operator by

$$
\hat n_i = c_i^\dagger c_i .
$$

Using the anticommutation relations, one can show that

$$
\hat n_i^2 = \hat n_i .
$$

Indeed,

$$
\hat n_i^2
=
c_i^\dagger c_i c_i^\dagger c_i .
$$

Using

$$
c_i c_i^\dagger = 1 - c_i^\dagger c_i ,
$$

we get

$$
\hat n_i^2
=
c_i^\dagger (1-c_i^\dagger c_i)c_i .
$$

Therefore

$$
\hat n_i^2
=
c_i^\dagger c_i
-
c_i^\dagger c_i^\dagger c_i c_i .
$$

But

$$
(c_i^\dagger)^2=0
$$

and

$$
c_i^2=0 .
$$

Hence the second term vanishes, and we obtain

$$
\hat n_i^2 = \hat n_i .
$$

If an operator satisfies

$$
\hat n_i^2=\hat n_i ,
$$

then its eigenvalues $\lambda$ must satisfy

$$
\lambda^2=\lambda .
$$

Thus

$$
\lambda(\lambda-1)=0 .
$$

Therefore

$$
\lambda = 0
$$

or

$$
\lambda = 1 .
$$

So the fermionic number operator has only the eigenvalues $0$ and $1$.

The ladder action is therefore

$$
c_i^\dagger |0_i\rangle = |1_i\rangle ,
$$

and

$$
c_i^\dagger |1_i\rangle = 0 .
$$

Similarly,

$$
c_i |1_i\rangle = |0_i\rangle ,
$$

and

$$
c_i |0_i\rangle = 0 .
$$

Thus

$$
c_i^\dagger c_i |0_i\rangle = 0,
$$

and

$$
c_i^\dagger c_i |1_i\rangle = |1_i\rangle .
$$

Hence

$$
\hat n_i |n_i\rangle = n_i |n_i\rangle ,
$$

with

$$
n_i\in\{0,1\}.
$$





## Representing fermionic states with bitstrings

Since the occupation numbers of fermionic modes can only be $0$ or $1$, we can represent the occupation of a system of $L$ fermionic modes as a bitstring of length $L$. Each bit in the string corresponds to a mode, with `0` indicating that the mode is unoccupied and `1` indicating that it is occupied. This is convenient for both theoretical and computational purposes, as it allows us to efficiently encode the state of the system and perform operations on it. We now define a python class to represent fermionic states using bitstrings.

```python
class FermionFockSpace:
    def __init__(self, num_modes):
        self.num_modes = num_modes

    def vacuum(self):
        return 0

    def occupation(self, state, site):
        return (state >> site) & 1

    def count_particles(self, state):
        return state.bit_count()

    def parity_before(self, state, site):
        mask = (1 << site) - 1
        num_before = (state & mask).bit_count()
        return -1 if num_before % 2 else 1

    def create(self, state, site):
        if self.occupation(state, site) == 1:
            return 0, None

        coeff = self.parity_before(state, site)
        new_state = state | (1 << site)

        return coeff, new_state

    def annihilate(self, state, site):
        if self.occupation(state, site) == 0:
            return 0, None

        coeff = self.parity_before(state, site)
        new_state = state & ~(1 << site)

        return coeff, new_state

    def number(self, state, site):
        return self.occupation(state, site), state

    def state_to_occupations(self, state):
        return [(state >> i) & 1 for i in range(self.num_modes)]

    def occupations_to_state(self, occupations):
        state = 0
        for i, n_i in enumerate(occupations):
            if n_i:
                state |= (1 << i)
        return state

    def pretty(self, state):
        occs = self.state_to_occupations(state)
        return "|" + ",".join(str(n) for n in occs) + ">"
    
```

```python
fs = FermionFockSpace(num_modes=4)

state = fs.vacuum()
print(fs.pretty(state))

coeff, state = fs.create(state, 1)
print(coeff, fs.pretty(state))

coeff, state = fs.create(state, 3)
print(coeff, fs.pretty(state))

coeff, new_state = fs.create(state, 1)
print(coeff, new_state)

coeff, state = fs.annihilate(state, 1)
print(coeff, fs.pretty(state))
```
