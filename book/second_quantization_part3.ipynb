# The Electronic Hamiltonian, Coulomb and Exchange Integrals, and CI Matrix Elements

## From operator body rank to molecular electronic structure

The nonrelativistic electronic Hamiltonian provides the most important application of one- and two-body operators. In these notes, we assemble the electronic Hamiltonian in second quantization and derive the structure of its matrix elements between Slater determinants.

The central result will be

$$
\boxed{
\hat H_{\mathrm{el}}
=
E_{\mathrm{NN}}
+
\sum_{pq}h_{pq}a_p^\dagger a_q
+
\frac{1}{4}
\sum_{pqrs}
\langle pq||rs\rangle
a_p^\dagger a_q^\dagger a_s a_r.
}
$$

From this form, the Slater--Condon rules and the sparse structure of configuration-interaction Hamiltonians follow naturally.

## 1. The fixed-nuclei electronic Hamiltonian in first quantization

Within the Born--Oppenheimer approximation, the nuclear positions \(\{\mathbf R_A\}\) are fixed. In atomic units, the electronic Hamiltonian for \(N_e\) electrons is

$$
\hat H_{\mathrm{el}}
=
E_{\mathrm{NN}}
+
\sum_{\alpha=1}^{N_e}
\hat h(\alpha)
+
\sum_{\alpha<\beta}
\frac{1}{r_{\alpha\beta}},
$$

where

$$
E_{\mathrm{NN}}
=
\sum_{A<B}
\frac{Z_AZ_B}{R_{AB}}
$$

is the nuclear-repulsion energy and

$$
\hat h(\alpha)
=
-\frac{1}{2}\nabla_\alpha^2
-
\sum_A
\frac{Z_A}{|\mathbf r_\alpha-\mathbf R_A|}
$$

is the one-electron core Hamiltonian.

The Hamiltonian separates by body rank:

$$
\boxed{
\begin{array}{ccl}
E_{\mathrm{NN}} &:& \text{zero-body scalar term},\\[4pt]
\displaystyle\sum_\alpha \hat h(\alpha) &:& \text{one-body electronic term},\\[8pt]
\displaystyle\sum_{\alpha<\beta} r_{\alpha\beta}^{-1} &:& \text{two-body electronic term}.
\end{array}
}
$$

No explicit three-electron interaction occurs in the ordinary Coulomb Hamiltonian.

## 2. Spin-orbitals and one-electron integrals

Let

$$
\chi_p(x)=\chi_p(\mathbf r,\omega)
$$

be orthonormal spin-orbitals, where \(\omega\) denotes the spin variable:

$$
\int dx\,\chi_p^\ast(x)\chi_q(x)=\delta_{pq}.
$$

The one-electron integral of the core Hamiltonian is

$$
\boxed{
h_{pq}
=
\int dx\,
\chi_p^\ast(x)
\left[
-\frac{1}{2}\nabla^2
-
\sum_A\frac{Z_A}{|\mathbf r-\mathbf R_A|}
\right]
\chi_q(x).
}
$$

It has a direct interpretation:

$$
h_{pq}
=
\text{amplitude for a one-electron transition }q\to p
\text{ induced by }\hat h.
$$

The one-body contribution in second quantization is therefore

$$
\boxed{
\hat H_1
=
\sum_{pq}
h_{pq}a_p^\dagger a_q.
}
$$

## 3. Two-electron integrals

For Coulomb repulsion,

$$
\hat g(1,2)=\frac{1}{r_{12}}.
$$

Using spin-orbitals, define

$$
\boxed{
(pq|rs)
=
\iint dx_1\,dx_2\,
\chi_p^\ast(x_1)\chi_q^\ast(x_2)
\frac{1}{r_{12}}
\chi_r(x_1)\chi_s(x_2).
}
$$

This integral describes a pair replacement

$$
(r,s)\longrightarrow(p,q).
$$

The antisymmetrized two-electron integral is

$$
\boxed{
\langle pq||rs\rangle
=
(pq|rs)-(pq|sr).
}
$$

The ordinary integral contains the direct pair matrix element. The antisymmetrized integral already combines the direct and exchange contributions required when the external states are fermionic determinants.

**Notation warning.** The index arrangement used here corresponds to the ordered spin-orbital matrix element \(\langle pq|r_{12}^{-1}|rs\rangle\). Some quantum-chemistry conventions print the same integral with a different index ordering. The integral definition, rather than the typography, should always be checked.

## 4. The electronic Hamiltonian in second quantization

The one-body and two-body mappings yield

$$
\boxed{
\hat H_{\mathrm{el}}
=
E_{\mathrm{NN}}
+
\sum_{pq}
h_{pq}a_p^\dagger a_q
+
\frac{1}{2}
\sum_{pqrs}
(pq|rs)
a_p^\dagger a_q^\dagger a_s a_r.
}
$$

Equivalently, using antisymmetrized integrals,

$$
\boxed{
\hat H_{\mathrm{el}}
=
E_{\mathrm{NN}}
+
\sum_{pq}
h_{pq}a_p^\dagger a_q
+
\frac{1}{4}
\sum_{pqrs}
\langle pq||rs\rangle
a_p^\dagger a_q^\dagger a_s a_r.
}
$$

The two representations are equivalent. The integral convention determines the prefactor:

$$
\boxed{
\begin{array}{ccl}
(pq|rs) &\text{with}& \displaystyle \frac{1}{2},\\[6pt]
\langle pq||rs\rangle &\text{with}& \displaystyle \frac{1}{4}.
\end{array}
}
$$

This Hamiltonian acts on every sector of Fock space, but in a calculation with a fixed electron number we restrict it to the desired \(N_e\)-electron sector.

## 5. Coulomb and exchange for two occupied orbitals

Let two electrons occupy spatial orbitals \(\varphi_i\) and \(\varphi_j\). For simplicity, first consider parallel spins so that the spin part allows exchange.

The Coulomb integral is

$$
\boxed{
J_{ij}
=
(ij|ij)
=
\iint d\mathbf r_1\,d\mathbf r_2\,
|\varphi_i(\mathbf r_1)|^2
\frac{1}{r_{12}}
|\varphi_j(\mathbf r_2)|^2.
}
$$

It resembles the classical electrostatic repulsion between the orbital charge distributions.

The exchange integral is

$$
\boxed{
K_{ij}
=
(ij|ji)
=
\iint d\mathbf r_1\,d\mathbf r_2\,
\varphi_i^\ast(\mathbf r_1)
\varphi_j^\ast(\mathbf r_2)
\frac{1}{r_{12}}
\varphi_j(\mathbf r_1)
\varphi_i(\mathbf r_2).
}
$$

For a same-spin fermionic pair, the interaction expectation value is

$$
\boxed{
\langle ij|\hat W^{(2)}|ij\rangle
=
(ij|ij)-(ij|ji)
=
J_{ij}-K_{ij}.
}
$$

Thus the exchange contribution lowers the repulsion between same-spin electrons in a determinant.

## 6. Why exchange vanishes between opposite-spin spin-orbitals

Let the spatial orbitals be combined with spin functions:

$$
\chi_{i\alpha}(x)=\varphi_i(\mathbf r)\alpha(\omega),
\qquad
\chi_{j\beta}(x)=\varphi_j(\mathbf r)\beta(\omega).
$$

Consider the exchange integral

$$
(i\alpha,j\beta|j\beta,i\alpha).
$$

Its spin part contains

$$
\langle\alpha|\beta\rangle
\langle\beta|\alpha\rangle.
$$

Since

$$
\langle\alpha|\beta\rangle=0,
$$

we obtain

$$
\boxed{
(i\alpha,j\beta|j\beta,i\alpha)=0.
}
$$

Hence an opposite-spin pair contributes only the Coulomb term, whereas a same-spin pair contributes direct minus exchange.

This is why the Hartree--Fock energy distinguishes same-spin and opposite-spin electron pairs even though the fundamental interaction is spin-independent Coulomb repulsion.

# Part I. Determinants and selection rules

## 7. Slater determinants in occupation-number form

Choose a fixed canonical ordering of spin-orbitals. A determinant with occupied orbitals

$$
i_1<i_2<\cdots<i_{N_e}
$$

is represented as

$$
|\Phi\rangle
=
a_{i_1}^\dagger
a_{i_2}^\dagger
\cdots
a_{i_{N_e}}^\dagger
|0\rangle.
$$

Every time a fermionic operator is moved through the ordered creation string, a minus sign is generated:

$$
a_p^\dagger a_q^\dagger=-a_q^\dagger a_p^\dagger.
$$

Thus determinant phases and all exchange signs are handled algebraically.

A singly replaced determinant is conveniently defined by

$$
\boxed{
|\Phi_i^a\rangle
=
a_a^\dagger a_i|\Phi\rangle,
}
$$

where \(i\) is occupied in \(|\Phi\rangle\) and \(a\) is unoccupied.

A doubly replaced determinant is defined by

$$
\boxed{
|\Phi_{ij}^{ab}\rangle
=
a_a^\dagger a_b^\dagger a_j a_i|\Phi\rangle.
}
$$

These operator definitions fix the relative phase convention of the excited determinants.

## 8. Selection rules from body rank

The Hamiltonian is

$$
\hat H_{\mathrm{el}}
=
E_{\mathrm{NN}}
+
\hat H_1
+
\hat H_2.
$$

The one-body term contains one annihilator and one creator:

$$
\hat H_1
=
\sum_{pq}h_{pq}a_p^\dagger a_q.
$$

Therefore it can change at most one occupied orbital.

The two-body term contains two annihilators and two creators:

$$
\hat H_2
=
\frac{1}{4}
\sum_{pqrs}
\langle pq||rs\rangle
a_p^\dagger a_q^\dagger a_s a_r.
$$

Therefore it can change at most two occupied orbitals.

Consequently,

$$
\boxed{
\langle\Phi_I|\hat H_{\mathrm{el}}|\Phi_J\rangle=0
\quad
\text{if }
|\Phi_I\rangle
\text{ and }
|\Phi_J\rangle
\text{ differ by more than two occupations.}
}
$$

This is not an approximation. It follows exactly from the two-body structure of the Coulomb Hamiltonian.

It is also the fundamental reason that configuration-interaction Hamiltonian matrices are sparse in a determinant basis.

# Part II. Slater--Condon matrix elements

## 9. Diagonal matrix element

Consider a determinant \(|\Phi\rangle\) with occupied spin-orbitals \(i,j,\ldots\).

For the one-body operator, only replacements that return the determinant to itself contribute:

$$
\langle\Phi|\hat H_1|\Phi\rangle
=
\sum_{i\in\mathrm{occ}}h_{ii}.
$$

For the two-body operator, both annihilated orbitals must be occupied and the created orbitals must restore the same determinant. The result is

$$
\langle\Phi|\hat H_2|\Phi\rangle
=
\frac{1}{2}
\sum_{i,j\in\mathrm{occ}}
\langle ij||ij\rangle.
$$

Therefore,

$$
\boxed{
\langle\Phi|\hat H_{\mathrm{el}}|\Phi\rangle
=
E_{\mathrm{NN}}
+
\sum_{i\in\mathrm{occ}}h_{ii}
+
\frac{1}{2}
\sum_{i,j\in\mathrm{occ}}
\langle ij||ij\rangle.
}
$$

Using

$$
\langle ij||ij\rangle
=
(ij|ij)-(ij|ji),
$$

this becomes

$$
\boxed{
\langle\Phi|\hat H_{\mathrm{el}}|\Phi\rangle
=
E_{\mathrm{NN}}
+
\sum_i h_{ii}
+
\frac{1}{2}\sum_{ij}
\left[
J_{ij}-K_{ij}
\right],
}
$$

where the sums run over occupied spin-orbitals. Terms with \(i=j\) vanish in the antisymmetrized expression because

$$
\langle ii||ii\rangle=0.
$$

## 10. Matrix element for a single replacement

Let

$$
|\Phi_i^a\rangle
=
a_a^\dagger a_i|\Phi\rangle,
$$

with occupied \(i\) replaced by unoccupied \(a\).

The one-body operator can directly make this replacement:

$$
\langle\Phi_i^a|\hat H_1|\Phi\rangle
=
h_{ai}.
$$

The two-body operator can also generate the same final determinant: it replaces \(i\to a\) while a second occupied spectator orbital \(j\) is removed and recreated. This contributes

$$
\sum_{j\in\mathrm{occ}}
\langle aj||ij\rangle.
$$

Hence,

$$
\boxed{
\langle\Phi_i^a|\hat H_{\mathrm{el}}|\Phi\rangle
=
h_{ai}
+
\sum_{j\in\mathrm{occ}}
\langle aj||ij\rangle.
}
$$

The scalar nuclear-repulsion term contributes only when bra and ket are identical and therefore does not appear here.

Define the Fock matrix associated with \(|\Phi\rangle\) by

$$
F_{pq}
=
h_{pq}
+
\sum_{j\in\mathrm{occ}}
\langle pj||qj\rangle.
$$

Then

$$
\boxed{
\langle\Phi_i^a|\hat H_{\mathrm{el}}|\Phi\rangle
=
F_{ai}.
}
$$

For canonical Hartree--Fock orbitals, occupied--virtual Fock matrix elements vanish:

$$
F_{ai}=0.
$$

This is Brillouin's theorem.

## 11. Matrix element for a double replacement

Let

$$
|\Phi_{ij}^{ab}\rangle
=
a_a^\dagger a_b^\dagger a_j a_i|\Phi\rangle.
$$

The determinant differs from \(|\Phi\rangle\) by two occupations:

$$
(i,j)\longrightarrow(a,b).
$$

A one-body operator cannot produce this change:

$$
\langle\Phi_{ij}^{ab}|\hat H_1|\Phi\rangle=0.
$$

The two-body operator has exactly the required replacement structure. Therefore,

$$
\boxed{
\langle\Phi_{ij}^{ab}|\hat H_{\mathrm{el}}|\Phi\rangle
=
\langle ab||ij\rangle.
}
$$

This matrix element is the antisymmetrized two-electron integral connecting the two determinants.

Thus the two-electron integral does not merely contribute to expectation values; it is also the direct coupling amplitude between determinants differing by a double excitation.

## 12. Determinants differing by three or more occupations

Suppose \(|\Phi_I\rangle\) and \(|\Phi_J\rangle\) differ in at least three occupied spin-orbitals.

The zero-body term changes no occupations.

The one-body term can change at most one occupation.

The two-body term can change at most two occupations.

Therefore no term in the ordinary electronic Hamiltonian can transform \(|\Phi_J\rangle\) into \(|\Phi_I\rangle\):

$$
\boxed{
\langle\Phi_I|\hat H_{\mathrm{el}}|\Phi_J\rangle=0
\qquad
\text{for differences of rank }3\text{ or higher}.
}
$$

This yields the Slater--Condon structure:

$$
\boxed{
\begin{array}{ccl}
0\text{ replacements} &:& \text{diagonal energy},\\[4pt]
1\text{ replacement} &:& \text{one-body term plus spectator pair terms},\\[4pt]
2\text{ replacements} &:& \text{antisymmetrized two-electron integral},\\[4pt]
>2\text{ replacements} &:& 0.
\end{array}
}
$$

# Part III. Concrete examples

## 13. Two electrons in two spatial orbitals: Coulomb versus exchange

Consider two spatial orbitals \(\varphi_i\) and \(\varphi_j\).

### Opposite-spin determinant

For the determinant

$$
|i\alpha,j\beta\rangle,
$$

the Coulomb contribution is

$$
J_{ij},
$$

while the exchange contribution vanishes because the spin functions are orthogonal:

$$
K_{i\alpha,j\beta}=0.
$$

Therefore,

$$
\boxed{
\langle i\alpha,j\beta|\hat W^{(2)}|i\alpha,j\beta\rangle
=
J_{ij}.
}
$$

### Same-spin determinant

For the determinant

$$
|i\alpha,j\alpha\rangle,
$$

the exchange term survives:

$$
\boxed{
\langle i\alpha,j\alpha|\hat W^{(2)}|i\alpha,j\alpha\rangle
=
J_{ij}-K_{ij}.
}
$$

The difference is not caused by a spin-dependent Coulomb interaction; it arises from antisymmetry of the electronic wavefunction.

## 14. Closed-shell two-electron example

Let \(i\) and \(a\) be two spatial orbitals. Consider the closed-shell determinants

$$
|\Phi_0\rangle
=
|i\alpha,i\beta\rangle
=
a_{i\alpha}^\dagger a_{i\beta}^\dagger|0\rangle,
$$

and

$$
|\Phi_{ii}^{aa}\rangle
=
|a\alpha,a\beta\rangle
=
a_{a\alpha}^\dagger a_{a\beta}^\dagger|0\rangle.
$$

The reference determinant energy is

$$
\boxed{
\langle\Phi_0|\hat H_{\mathrm{el}}|\Phi_0\rangle
=
E_{\mathrm{NN}}
+
2h_{ii}
+
(ii|ii),
}
$$

because the two electrons have opposite spin and the exchange contribution between \(i\alpha\) and \(i\beta\) vanishes.

Similarly,

$$
\boxed{
\langle\Phi_{ii}^{aa}|\hat H_{\mathrm{el}}|\Phi_{ii}^{aa}\rangle
=
E_{\mathrm{NN}}
+
2h_{aa}
+
(aa|aa).
}
$$

The coupling between the two doubly occupied configurations is

$$
\begin{aligned}
\langle a\alpha,a\beta|\hat H_{\mathrm{el}}|i\alpha,i\beta\rangle
&=
\langle a\alpha\,a\beta||i\alpha\,i\beta\rangle\\
&=
(a\alpha\,a\beta|i\alpha\,i\beta),
\end{aligned}
$$

because the exchanged spin contribution vanishes. In spatial-orbital notation this is the Coulomb-like pair-transfer integral

$$
\boxed{
\langle a\alpha,a\beta|\hat H_{\mathrm{el}}|i\alpha,i\beta\rangle
=
(aa|ii).
}
$$

This provides a minimal illustration of how a two-electron integral couples configurations in a CI Hamiltonian.

# Part IV. From determinants to correlated wavefunctions

## 15. Configuration interaction matrix elements

A configuration-interaction wavefunction is expanded in determinants or spin-adapted configuration state functions:

$$
|\Psi\rangle
=
\sum_I C_I|\Phi_I\rangle.
$$

The Schrödinger equation becomes the matrix eigenvalue problem

$$
\sum_J
\langle\Phi_I|\hat H_{\mathrm{el}}|\Phi_J\rangle
C_J
=
E C_I.
$$

The matrix elements are assembled from one- and two-electron integrals according to the replacement rank between \(\Phi_I\) and \(\Phi_J\).

Because the electronic Hamiltonian contains at most two-body terms,

$$
H_{IJ}=0
$$

whenever the determinants differ by more than two spin-orbitals.

Thus:

$$
\boxed{
\begin{array}{c}
\text{orbital integrals from first quantization}\\[4pt]
\Downarrow\\[4pt]
\text{second-quantized replacement operators}\\[4pt]
\Downarrow\\[4pt]
\text{Slater--Condon matrix elements}\\[4pt]
\Downarrow\\[4pt]
\text{CI Hamiltonian matrix}.
\end{array}
}
$$

## 16. Expectation values and reduced density matrices

For a general correlated state \(|\Psi\rangle\), define the one-particle reduced density matrix

$$
\boxed{
\gamma_{qp}
=
\langle\Psi|a_p^\dagger a_q|\Psi\rangle,
}
$$

and the two-particle reduced density matrix

$$
\boxed{
\Gamma_{rs,pq}
=
\langle\Psi|
a_p^\dagger a_q^\dagger a_s a_r
|\Psi\rangle.
}
$$

Then the electronic energy can be written directly as

$$
\boxed{
E
=
E_{\mathrm{NN}}
+
\sum_{pq}h_{pq}\gamma_{qp}
+
\frac{1}{2}
\sum_{pqrs}
(pq|rs)\Gamma_{rs,pq}.
}
$$

Equivalently, using antisymmetrized integrals with the appropriate two-particle density convention,

$$
E
=
E_{\mathrm{NN}}
+
\sum_{pq}h_{pq}\gamma_{qp}
+
\frac{1}{4}
\sum_{pqrs}
\langle pq||rs\rangle\Gamma_{rs,pq}.
$$

This reveals a general principle:

$$
\boxed{
\text{A \(k\)-body observable requires at most the \(k\)-particle reduced density matrix.}
}
$$

Thus the same operator structure that determines determinant couplings also determines which reduced densities are required to compute observables of correlated states.

## 17. Complete summary

The first-quantized fixed-nuclei electronic Hamiltonian is

$$
\hat H_{\mathrm{el}}
=
E_{\mathrm{NN}}
+
\sum_{\alpha}\hat h(\alpha)
+
\sum_{\alpha<\beta}\frac{1}{r_{\alpha\beta}}.
$$

Its orbital matrix elements are

$$
h_{pq}
=
\langle\chi_p|\hat h|\chi_q\rangle,
$$

and

$$
(pq|rs)
=
\langle\chi_p\chi_q|
r_{12}^{-1}
|\chi_r\chi_s\rangle.
$$

Second quantization transforms these into replacement operators:

$$
\boxed{
\hat H_{\mathrm{el}}
=
E_{\mathrm{NN}}
+
\sum_{pq}h_{pq}a_p^\dagger a_q
+
\frac{1}{2}
\sum_{pqrs}(pq|rs)
a_p^\dagger a_q^\dagger a_s a_r.
}
$$

For fermions,

$$
\langle pq|\hat H_2|rs\rangle
=
(pq|rs)-(pq|sr)
=
\langle pq||rs\rangle,
$$

so equivalently

$$
\boxed{
\hat H_{\mathrm{el}}
=
E_{\mathrm{NN}}
+
\sum_{pq}h_{pq}a_p^\dagger a_q
+
\frac{1}{4}
\sum_{pqrs}\langle pq||rs\rangle
a_p^\dagger a_q^\dagger a_s a_r.
}
$$

The determinant matrix elements are:

$$
\boxed{
\langle\Phi|\hat H|\Phi\rangle
=
E_{\mathrm{NN}}
+
\sum_i h_{ii}
+
\frac{1}{2}\sum_{ij}\langle ij||ij\rangle,
}
$$

$$
\boxed{
\langle\Phi_i^a|\hat H|\Phi\rangle
=
h_{ai}
+
\sum_j\langle aj||ij\rangle,
}
$$

$$
\boxed{
\langle\Phi_{ij}^{ab}|\hat H|\Phi\rangle
=
\langle ab||ij\rangle,
}
$$

and

$$
\boxed{
\langle\Phi_I|\hat H|\Phi_J\rangle=0
\quad
\text{if the determinants differ by more than two occupations.}
}
$$

This is the route from the known one- and two-electron integrals of first quantization to electronic-structure Hamiltonian matrices in a determinant or CI basis.
