# Symbolic generation of nuclear attraction integral code

The next type of integrals we shall examine after implementing overlaps and kinetic energy integrals are the nuclear attraction integrals. 

A nuclear attraction integral between two basis functions $g_{ijk}$ and $g_{lmn}$ centered at $\vec A$ and $\vec B$, respectively, and with respect to a nucleus centered at $\vec C$ is defined as

$$
V_{A,B}^{ijk,lmn}(C)
=\int g_{ijk}(\vec r;\alpha,\vec A)\,\frac{Z_C}{r_C}\,g_{lmn}(\vec r;\beta,\vec B)\,d^3r,
\qquad r_C:=\|\vec r-\vec C\|.
$$

Here, $Z_C$ represents the nuclear charge of an atom at $C$. Just like overlap integrals, we shall at first calculate the nuclear attraction integrals between two $s$-orbitals, i.e. $(i,j,k)=(l,m,n)=(0,0,0)$ which will give us the base case integral. 

For notational simplicity, we will define

$$
V_p(C):=V_{A,B}^{000,000}(C).
$$

Integrals involving higher angular momenta can then be obtained again by applying recursive relations of Hermite Gaussians.
Unlike the overlap and kinetic energy integrals, nuclear attraction integrals are not factorizable into one-dimensional integrals because of the factor $1/r_C$ that couples all three cartesian coordinates of an electron. Thus, we will introduce some useful tricks that will allow us to deal with type of integrals that involve Coulomb interaction terms.

## Evaluation of $V_{A,B}^{000,000}(C)$

The basic idea that we are using in order to analytically calculate the nuclear repulsion integral is to transform the difficult
term $1/r_C$ into something that can be reduced to a Gaussian integral. We start by recalling the analytic
solution of a 1D Gaussian integral: 

$$
\int_{-\infty}^{\infty} e^{-\alpha x^2}\,dx=\sqrt{\frac{\pi}{\alpha}}.
$$

If we choose $\alpha=r_C^2$ (and change the integration variable to $t$), we can express this as

$$
\int_{-\infty}^{\infty} e^{-r_C^2 t^2}\,dt=\sqrt{\frac{\pi}{r_C^2}}=\frac{\sqrt{\pi}}{r_C}.
$$

This means we can express the Coulomb $1/r_C$ factor as a Gaussian integral over additional auxiliary variable

$$
\frac{1}{r_C}
=\frac{1}{\sqrt{\pi}}\int_{-\infty}^{\infty} e^{-r_C^2 t^2}\,dt
=\frac{2}{\sqrt{\pi}}\int_{0}^{\infty} e^{-r_C^2 t^2}\,dt.
$$

With this transformation we are effectively representing the Coulomb term as a (continous) superposition of Gaussians!


Substituting this into the nuclear attraction integral, we obtain 

$$
V_p(C)
=\frac{Z_C}{\sqrt{\pi}}
\int g_{000}(\vec r;\alpha,\vec A)\,g_{000}(\vec r;\beta,\vec B)
\left(\int_{-\infty}^{\infty} e^{-r_C^2 t^2}\,dt\right)\,d^3r.
$$

We have thus transformed the original 3-dimensional integral into a 4-dimensional integral, which at the first look seems to even more complicate the things. However, the new integral does not have an annoying $1/r_C$ factor and is now separable (for fixed $t$ it is just a product of Gaussians in $\vec r$).

Since we now deal with a product of Gaussians we can apply the Gaussian product theorem (GPT) to 
$g_{000}(\vec r;\alpha,\vec A)\,g_{000}(\vec r;\beta,\vec B)$, which gives us just 

$$
\exp(-\mu R_{AB}^2)\,g_{000}(\vec r;p,\vec P)
$$ 

with the following abbreviations:

$$
p=\alpha+\beta,\qquad
\vec P=\frac{\alpha\vec A+\beta\vec B}{p},\qquad
\mu=\frac{\alpha\beta}{\alpha+\beta},\qquad
R_{AB}=\|\vec A-\vec B\|.
$$

The nuclear attraction integral can be thus reduced to the form

$$
V_p(C)
=\frac{Z_C}{\sqrt{\pi}}\exp(-\mu R_{AB}^2)\int_{-\infty}^{\infty}\int
e^{-p\|\vec r-\vec P\|^2}\,e^{-t^2\|\vec r-\vec C\|^2}\,d^3r\,dt.
$$

Now we can apply Gaussian product theorem one more time to the Gaussians that depend on $\vec r$:

$$
e^{-p\|\vec r-\vec P\|^2}\,e^{-t^2\|\vec r-\vec C\|^2}
=\exp\!\left(-\frac{p\,t^2}{p+t^2}R_{PC}^2\right)\,
\exp\!\left(-(p+t^2)\|\vec r-\vec S\|^2\right),
$$

with further abbreviations 

$$
R_{PC}=\|\vec P-\vec C\|,\qquad
\vec S=\frac{p\vec P+t^2\vec C}{p+t^2}.
$$

The vector $\vec S$ is just like $\vec P$ for our first product gaussian, but its exact form is not important, because it just represents a constant translation in $\vec r$ and we are going to integrate over the whole space anyway.



Thus,
$$
V_p(C)
=\frac{Z_C}{\sqrt{\pi}}\exp(-\mu R_{AB}^2)\int_{-\infty}^{\infty}
\exp\!\left(-\frac{p\,t^2}{p+t^2}R_{PC}^2\right)
\left(\int \exp\!\left(-(p+t^2)\|\vec r-\vec S\|^2\right)\,d^3r\right)\,dt.
$$

The integral over $\vec r$ is now just a 3D Gaussian integral with a well known value:
$$
\int \exp\!\left(-(p+t^2)\|\vec r-\vec S\|^2\right)\,d^3r
=\left(\frac{\pi}{p+t^2}\right)^{3/2}.
$$



By inserting it, the whole problem reduces to solving a 1D integral:

$$
V_p(C)
=\frac{Z_C}{\sqrt{\pi}}\exp(-\mu R_{AB}^2)\int_{-\infty}^{\infty}
\left(\frac{\pi}{p+t^2}\right)^{3/2}
\exp\!\left(-\frac{p\,t^2}{p+t^2}R_{PC}^2\right)\,dt.
$$

In order to proceed we first exploit the symmetry of the remaining integrand to change the integration limits to $0$ and $\infty$:

$$
V_p(C)
=\frac{2Z_C}{\sqrt{\pi}}\exp(-\mu R_{AB}^2)\int_{0}^{\infty}
\left(\frac{\pi}{p+t^2}\right)^{3/2}
\exp\!\left(-\frac{p\,t^2}{p+t^2}R_{PC}^2\right)\,dt.
$$

Furthermore, we introduce a new variable

$$
u^2=\frac{t^2}{p+t^2}
\qquad\Longleftrightarrow\qquad
t^2=\frac{p\,u^2}{1-u^2},
\qquad
p+t^2=\frac{p}{1-u^2}.
$$

Differentiating $t=\sqrt{p}\,u(1-u^2)^{-1/2}$ gives

$$
dt=\sqrt{p}\,(1-u^2)^{-3/2}\,du.
$$



Upon substitution, the integral becomes

$$
V_p(C)
=\frac{2Z_C}{\sqrt{\pi}}\exp(-\mu R_{AB}^2)\int_{0}^{1}
\left(\frac{\pi}{p/(1-u^2)}\right)^{3/2}
\exp(-p\,u^2R_{PC}^2)\,\sqrt{p}(1-u^2)^{-3/2}\,du.
$$

The $(1-u^2)$ factors cancel exactly leading to the following expression:

$$
V_p(C)
=\frac{2Z_C}{\sqrt{\pi}}\exp(-\mu R_{AB}^2)\,\frac{\pi^{3/2}}{p}\int_{0}^{1}\exp(-p\,u^2R_{PC}^2)\,du
=\frac{2\pi Z_C}{p}\exp(-\mu R_{AB}^2)\int_{0}^{1}\exp(-p\,u^2R_{PC}^2)\,du.
$$



The remaining integral, unfortunately, cannot be solved in elementary functions in general. This integral 
defines a special function that is in the field of quantum chemistry usually called Boys function (after Francis Boys, a pioneer of the field):

$$
F_n(x):=\int_{0}^{1}\exp(-x t^2)\,t^{2n}\,dt.
$$

Expressed in terms of the Boys function, the nuclear attraction integral becomes

$$
V_p(C)=\frac{2\pi Z_C}{p}\exp(-\mu R_{AB}^2)\,F_0\!\left(pR_{PC}^2\right).
$$

Although the last expression we derived seemingly does not involve the integration, the Boys function is still defined as an integral. We have effectively reduced a 3-dimensional integral over the whole space to a 1-dimensional integral over the interval $[0,1]$. This is a huge simplification, since such 1D integrals can be efficiently approximated using many numerical techniques!

## Evaluation of arbitrary nuclear attraction integrals

We can now obtain the nuclear attraction integral between Hermite Gaussians of arbitrary angular momenta $W_{A,B}^{ijk,lmn}(C)$ by utilizing its definition through derivatives:
$$
W_{A,B}^{ijk,lmn}(C)
=\int h_{ijk}(\vec r;\alpha,\vec A)\,\frac{Z_C}{r_C}\,h_{lmn}(\vec r;\beta,\vec B)\,d^3r
=\left(\frac{\partial}{\partial A_x}\right)^i\left(\frac{\partial}{\partial A_y}\right)^j\left(\frac{\partial}{\partial A_z}\right)^k
\left(\frac{\partial}{\partial B_x}\right)^l\left(\frac{\partial}{\partial B_y}\right)^m\left(\frac{\partial}{\partial B_z}\right)^n
V_p(C).
$$
(Depending on the sign convention used in defining the Hermite Gaussians via center-derivatives, an overall factor of $(-1)^{l+m+n}$ may appear; the workflow is unchanged.)

The dependency on nuclear coordinates $\vec A$ and $\vec B$ comes from the exponential part as well as the Boys function. While we already know how to differentiate the exponential part, we still need to find derivatives of the Boys function. We start by taking the derivative of the Boys function directly using its definition:
$$
\frac{d}{dx}F_n(x)
=\frac{d}{dx}\int_{0}^{1}\exp(-xt^2)\,t^{2n}\,dt
=\int_{0}^{1}\frac{d}{dx}\Big(\exp(-xt^2)\Big)\,t^{2n}\,dt
=\int_{0}^{1}(-t^2)\exp(-xt^2)\,t^{2n}\,dt
=-\int_{0}^{1}\exp(-xt^2)\,t^{2(n+1)}\,dt
=-F_{n+1}(x).
$$

Although we were not able to get something simpler through differentiation, at least the expression does not get more complicated and we still only have to evaluate Boys functions numerically.

Now we can differentiate $V_p(C)$ with respect to $A_x$ and $B_x$ according to the desired angular momenta to obtain arbitrary nuclear attraction integrals between Hermite Gaussians. To obtain nuclear attraction integrals between Cartesian Gaussians, $V_{A,B}^{ijk,lmn}(C)$, we simply use Hermite Gaussian expansion:
$$
V_{A,B}^{ijk,lmn}(C)
=\sum_{a b c}\sum_{d e f}
c_a^{\,i}\,c_b^{\,j}\,c_c^{\,k}\,c_d^{\,l}\,c_e^{\,m}\,c_f^{\,n}\;
W_{A,B}^{abc,def}(C).
$$

Although we have now derived a general expression for nuclear attraction integrals, the expressions are everything but simple. 
In the following we again use the symbolic algebra as implemented in sympy in order to generate analytic expressions for the integrals that will be subsequently writen to an automatically generated Python code. 

```python
import sympy as sp
```

```python
x, y, z = sp.symbols('x y z', real=True)
alpha, beta = sp.symbols("alpha beta", positive=True, real=True)
Ax, Ay, Az, Bx, By, Bz, Cx, Cy, Cz = sp.symbols("Ax Ay Az Bx By Bz Cx Cy Cz", real=True)
```

We implement provisionally `Boys` as a custom SymPy `Function` so that it behaves as a symbolic placeholder during derivations/code generation.

- `eval` returns `None` so SymPy keeps `boys(n, x)` unevaluated (no forced closed form).
- `fdiff` encodes the key analytic rule used in recurrences:

  $$
  \frac{d}{dx}F_n(x) = -F_{n+1}(x)
  $$

- This lets SymPy differentiate large integral expressions correctly while keeping them compact and structured.

```python
class Boys(sp.Function):
    nargs = 2

    def fdiff(self, argindex=1):
        if argindex == 2:
            return -Boys(self.args[0] + 1, self.args[1])
        raise ValueError(argindex)
```

```python
p = alpha + beta
q = alpha * beta
rab2 = (Ax - Bx) ** 2 + (Ay - By) ** 2 + (Az - Bz) ** 2
Px = (alpha * Ax + beta * Bx) / p
Py = (alpha * Ay + beta * By) / p
Pz = (alpha * Az + beta * Bz) / p
t_arg = p * ((Px - Cx) ** 2 + (Py - Cy) ** 2 + (Pz - Cz) ** 2)
```

```python
V00 = 2 * sp.pi * sp.exp(-q * rab2 / p) * Boys(0, t_arg) / p
V00
```

```python
from functools import lru_cache
@lru_cache(maxsize=None)
def get_ckn(k: int, n: int, p):
    """Coefficient C_k^n in the Cartesian-to-Hermite expansion."""
    if k < 0 or k > n:
        return sp.Integer(0)

    if n == 0:
        return sp.Integer(1) if k == 0 else sp.Integer(0)

    return get_ckn(k - 1, n - 1, p) / (2 * p) + (k + 1) * get_ckn(k + 1, n - 1, p)
```

```python
from functools import lru_cache
def build_derivatives_table(base_integral, Lmax, Avars, Bvars):
    Ax, Ay, Az = Avars
    Bx, By, Bz = Bvars

    all_idx = [
        (i, j, L - i - j)
        for L in range(Lmax + 1)
        for i in range(L + 1)
        for j in range(L + 1 - i)
    ]

    @lru_cache(maxsize=None)
    def derivative(i, j, k, l, m, n):
        if (i, j, k, l, m, n) == (0, 0, 0, 0, 0, 0):
            return base_integral

        if i > 0:
            return sp.diff(derivative(i - 1, j, k, l, m, n), Ax)
        if j > 0:
            return sp.diff(derivative(i, j - 1, k, l, m, n), Ay)
        if k > 0:
            return sp.diff(derivative(i, j, k - 1, l, m, n), Az)
        if l > 0:
            return sp.diff(derivative(i, j, k, l - 1, m, n), Bx)
        if m > 0:
            return sp.diff(derivative(i, j, k, l, m - 1, n), By)
        return sp.diff(derivative(i, j, k, l, m, n - 1), Bz)

    derivatives_dict = {}
    for (i, j, k) in all_idx:
        for (l, m, n) in all_idx:
            derivatives_dict[(i, j, k, l, m, n)] = sp.simplify(derivative(i, j, k, l, m, n))     

    return derivatives_dict
```

```python
Lmax = 1
```

```python
derivatives_dict = build_derivatives_table(V00, Lmax, (Ax, Ay, Az), (Bx, By, Bz))
```

```python
def get_integral_expressions(integral_indices, derivatives_dict, alpha, beta):
    integral_expressions = {}

    for (i,j,k,l,m,n) in integral_indices:

        ci = [get_ckn(o, i, alpha) for o in range(i + 1)]
        cj = [get_ckn(p, j, alpha) for p in range(j + 1)]
        ck = [get_ckn(q, k, alpha) for q in range(k + 1)]

        cl = [get_ckn(r, l, beta) for r in range(l + 1)]
        cm = [get_ckn(s, m, beta) for s in range(m + 1)]
        cn = [get_ckn(t, n, beta) for t in range(n + 1)]

        expr = 0
        for o, co in enumerate(ci):
            for p, cp in enumerate(cj):
                for q, cq in enumerate(ck):
                    for r, cr in enumerate(cl):
                        for s, cs in enumerate(cm):
                            for t, ct in enumerate(cn):
                                expr += (
                                    co * cp * cq * cr * cs * ct
                                    * derivatives_dict[(o, p, q, r, s, t)]
                                )

        integral_expressions[(i,j,k,l,m,n)] = sp.simplify(expr)
    return integral_expressions
```

### Which indices are generated?

We will generate all possible index tuples $(i,j,k,l,m,n)$ for Cartesian Gaussians up to angular momentum $L_{\max}=3$, which means that we will generate formulas for all s-, p-, d-, and f-type Cartesian Gaussians. This is not the most efficient approach since it ignores the symmetry and is redundant since many of the integrals are related by permutation of arguments. The length of the generated code could be reduced by generating only the unique integrals and then applying symmetry relations to obtain the rest. However, for simplicity and clarity we will generate all integrals explicitly. Later on this could be optimized by generating only the unique integrals and applying symmetry relations to obtain the rest. 

We implement a helper function l_to_ijk(L) that generates the index tuples for a given value of L, and then we use it to generate all index tuples up to Lmax.

```python
def l_to_ijk(L):
    IJK = []
    for I in range(L, -1, -1):
        for J in range(L - I, -1, -1):
            IJK.append((I, J, L - I - J))
    return sorted(IJK, reverse=True)

integral_indices=[]
ijk = [t for L in range(Lmax+1) for t in l_to_ijk(L)]
for i in ijk:
    for j in ijk:
        integral_indices.append(i+j)
integral_indices
```

### Use canonical representative indices

```python
from itertools import permutations
La = 0
Lb = 1
ijk = l_to_ijk(La)
lmn = l_to_ijk(Lb)
S3 = list(permutations([0, 1, 2]))

ijklmn_to_rep = {}
for i in ijk:
    for j in lmn:
        orbit = set()
        for perm in S3:
            i_perm = tuple(i[idx] for idx in perm)
            j_perm = tuple(j[idx] for idx in perm)
            orbit.add(i_perm + j_perm)
        representative = min(orbit) 
        ijklmn_to_rep.update({idx: representative for idx in orbit})
ijklmn_to_rep
```

```python
integral_expressions = get_integral_expressions(integral_indices, derivatives_dict, alpha, beta)
```

```python
for key, value in integral_expressions.items():
    print(f"Integral {key}: {value}")
```

### Simplify expressions using substitutions 

```python
# compact symbols
p, mu = sp.symbols("p mu", positive=True, real=True)
ABx, ABy, ABz, RAB2 = sp.symbols("ABx ABy ABz RAB2", real=True)
Px, Py, Pz = sp.symbols("Px Py Pz", real=True)
Qx, Qy, Qz = sp.symbols("Qx Qy Qz", real=True)      # Q = alpha*A + beta*B - p*C
PCx, PCy, PCz = sp.symbols("PCx PCy PCz", real=True)
u = sp.symbols("u", nonnegative=True, real=True)    # Boys argument
KAB = sp.symbols("KAB", positive=True, real=True)

subsdict = {
    # basic Gaussian combinations
    alpha + beta: p,
    alpha*beta/(alpha + beta): mu,

    # A-B differences
    Ax - Bx: ABx,
    Ay - By: ABy,
    Az - Bz: ABz,
    (Ax - Bx)**2 + (Ay - By)**2 + (Az - Bz)**2: RAB2,

    # product center P
    (alpha*Ax + beta*Bx)/(alpha + beta): Px,
    (alpha*Ay + beta*By)/(alpha + beta): Py,
    (alpha*Az + beta*Bz)/(alpha + beta): Pz,

    # Q = p*(P-C)
    alpha*Ax + beta*Bx - Cx*(alpha + beta): Qx,
    alpha*Ay + beta*By - Cy*(alpha + beta): Qy,
    alpha*Az + beta*Bz - Cz*(alpha + beta): Qz,

    # optional normalized Q-components
    Qx/p: PCx,
    Qy/p: PCy,
    Qz/p: PCz,

    # Boys argument
    ((alpha*Ax + beta*Bx - Cx*(alpha + beta))**2
     + (alpha*Ay + beta*By - Cy*(alpha + beta))**2
     + (alpha*Az + beta*Bz - Cz*(alpha + beta))**2)/(alpha + beta): u,

    # Gaussian prefactor
    sp.exp(-alpha*beta*((Ax - Bx)**2 + (Ay - By)**2 + (Az - Bz)**2)/(alpha + beta)): KAB,
}
```

```python
for key, value in integral_expressions.items():
    integral_expressions[key] = sp.simplify(value.subs(subsdict, simultaneous=True))
for key, value in integral_expressions.items():
    print(f"Integral {key}: {value}")
```

## Generate code for the overlap integrals

```python
from pathlib import Path
from sympy.printing.numpy import NumPyPrinter, _known_functions_numpy, _known_constants_numpy

class TheochemNumPyPrinter(NumPyPrinter):
    def print_boys(self, expr):
        n, t = expr.args
        return f"boys({self._print(n)}, {self._print(t)})"

printer = TheochemNumPyPrinter()
printer._module = "np"
printer.known_functions = {k: f"np.{v}" for k, v in _known_functions_numpy.items()}
printer.known_functions["Boys"] = "Boys"
printer.known_constants = {k: f"np.{v}" for k, v in _known_constants_numpy.items()}
```

```python
def write_v_module(path, name="V", use_cse=False, integral_expressions=None):
    lines = ["import numpy as np", 
             "from numba import njit",
             "@njit(cache=True, fastmath=True)",
             f"def {name}(i, j, k, l, m, n, ABx, ABy, ABz,p,alpha, beta,Qx, Qy, Qz,u,KAB):    "]

    for key, value in integral_expressions.items():
        lines.append(f"    if (i, j, k, l, m, n) == {key}:")
        if use_cse:
            repls, reduced = sp.cse(value, symbols=sp.numbered_symbols("t"))
            for sym, expr in repls:
                lines.append(f"        {sym} = {printer.doprint(expr)}")
            lines.append(f"        return {printer.doprint(reduced[0])}")
        else:
            lines.append(f"        return {printer.doprint(value)}")

    lines.append("    raise KeyError((i, j, k, l, m, n))")

    with open(path, "w", encoding="utf-8") as f:
        f.write("\n".join(lines) + "\n")
  
write_v_module(Path("/Users/rolandmitric/WORK/GITHUB/master_programming_2026/src/theochem2026/integrals/V.py"), name="V", use_cse = True, integral_expressions=integral_expressions)
```
