# Symbolic generation of electron repulsion integrals

The final and most expensive class of two-electron quantities are the electron-repulsion integrals (ERIs):

$$
(\mu\nu|\lambda\sigma)
=
\int d^3 r_1 \int d^3 r_2\,
\phi_\mu(\mathbf r_1)\phi_\nu(\mathbf r_1)\frac{1}{r_{12}}\phi_\lambda(\mathbf r_2)\phi_\sigma(\mathbf r_2),
\qquad r_{12}=|\mathbf r_1-\mathbf r_2|.
$$

As with overlap and nuclear-attraction integrals, we first derive the basic $s$-type case and then obtain higher angular momenta through the calculation of derivatives and the Hermite- Gaussian recursions.

For compact notation we define the basis integral as:

$$
G_{ABCD}^{000,000,000,000}=\langle ss|ss\rangle.
$$


### Evaluation of $\langle ss|ss\rangle$

For four primitive $s$-Gaussians the electron-repulsion integral is given by

$$
\langle ss|ss\rangle
=
\int d^3 r_1 \int d^3 r_2\,
e^{-\alpha|\mathbf r_1-\mathbf A|^2}
e^{-\beta|\mathbf r_1-\mathbf B|^2}
\frac{1}{r_{12}}
e^{-\gamma|\mathbf r_2-\mathbf C|^2}
e^{-\delta|\mathbf r_2-\mathbf D|^2}.
$$

We can apply the Gaussian product theorem separately to the electron-1 and electron-2 pairs, which gives us the following definitions for the combined exponents, Gaussian centers, and prefactors:

$$
p=\alpha+\beta,
\qquad
\mu=\frac{\alpha\beta}{p},
\qquad
\mathbf P=\frac{\alpha\mathbf A+\beta\mathbf B}{p},
$$

$$
q=\gamma+\delta,
\qquad
\eta=\frac{\gamma\delta}{q},
\qquad
\mathbf Q=\frac{\gamma\mathbf C+\delta\mathbf D}{q}.
$$

The basic $s$-type ERI can then be written as

$$
\langle ss|ss\rangle
=e^{-\mu R_{AB}^2}e^{-\eta R_{CD}^2}
\int d^3 r_1 \int d^3 r_2\,
e^{-p|\mathbf r_1-\mathbf P|^2}e^{-q|\mathbf r_2-\mathbf Q|^2}\frac{1}{r_{12}},
$$

with $R_{AB}=|\mathbf A-\mathbf B|$ and $R_{CD}=|\mathbf C-\mathbf D|$.


As for nuclear attraction, we remove the Coulomb denominator using the Gaussian transform:

$$
\frac{1}{r_{12}}=\frac{2}{\sqrt{\pi}}\int_0^{\infty}e^{-t^2 r_{12}^2}\,dt.
$$

Substituting gives

$$
\langle ss|ss\rangle
=\frac{2}{\sqrt{\pi}}e^{-\mu R_{AB}^2}e^{-\eta R_{CD}^2}
\int_0^{\infty} J(t)\,dt,
$$

where

$$
J(t)=
\int d^3 r_1 \int d^3 r_2\,
e^{-p|\mathbf r_1-\mathbf P|^2}
e^{-q|\mathbf r_2-\mathbf Q|^2}
e^{-t^2|\mathbf r_1-\mathbf r_2|^2}.
$$

For fixed $t$, this is a Gaussian integral in both electron coordinates.


We can first combine the $\mathbf r_1$-dependent Gaussians into a single Gaussian using the Gaussian product theorem:

$$
e^{-p|\mathbf r_1-\mathbf P|^2}e^{-t^2|\mathbf r_1-\mathbf r_2|^2}
=
\exp\left[-\frac{pt^2}{p+t^2}|\mathbf P-\mathbf r_2|^2\right]
\exp\left[-(p+t^2)|\mathbf r_1-\mathbf S|^2\right].
$$

The $\mathbf r_1$ integral is then a standard Gaussian integral, and we have    

$$
\int d^3 r_1\, e^{-p|\mathbf r_1-\mathbf P|^2}e^{-t^2|\mathbf r_1-\mathbf r_2|^2}
=\left(\frac{\pi}{p+t^2}\right)^{3/2}
\exp\left[-\frac{pt^2}{p+t^2}|\mathbf P-\mathbf r_2|^2\right].
$$

The remaining $\mathbf r_2$ integral is again Gaussian, and after a second GPT application:

$$
J(t)=\frac{\pi^3}{\left[pq+(p+q)t^2\right]^{3/2}}
\exp\left[-\frac{pq\,t^2}{pq+(p+q)t^2}R_{PQ}^2\right],
$$

with $R_{PQ}=|\mathbf P-\mathbf Q|$.


Inserting $J(t)$ into the integral over $t$ gives:

$$
\langle ss|ss\rangle
=\frac{2\pi^3}{\sqrt{\pi}}e^{-\mu R_{AB}^2}e^{-\eta R_{CD}^2}
\int_0^{\infty}
\frac{\exp\left[-\frac{pq\,t^2}{pq+(p+q)t^2}R_{PQ}^2\right]}{\left[pq+(p+q)t^2\right]^{3/2}}\,dt.
$$

This can be brought into a more compact form by defining

$$
\nu^2=\frac{(p+q)t^2}{pq+(p+q)t^2},
\qquad
\rho=\frac{pq}{p+q}.
$$

This transforms the integral to

$$
\langle ss|ss\rangle
=\frac{2\pi^{5/2}}{pq\sqrt{p+q}}e^{-\mu R_{AB}^2}e^{-\eta R_{CD}^2}
\int_0^1 e^{-\rho R_{PQ}^2\nu^2}\,d\nu.
$$

Using

$$
F_0(x)=\int_0^1 e^{-x t^2}\,dt,
$$

we obtain

$$
\langle ss|ss\rangle
=\frac{2\pi}{\rho}\left(\frac{\pi}{p+q}\right)^{3/2}
\exp\left(-\frac{\alpha\beta}{p}R_{AB}^2\right)
\exp\left(-\frac{\gamma\delta}{q}R_{CD}^2\right)
F_0\left(\rho R_{PQ}^2\right).
$$

This is the final expression for the basic $s$-type ERI. Higher angular momenta can be obtained through differentiation and recursion, as we will see in the next section.


## Evaluation of arbitrary ERIs

For Hermite Gaussians on centers $A,B,C,D$, define
$$
W_{ABCD}^{ijk,lmn,pqr,stu}
=
\int d^3 r_1 \int d^3 r_2\,
h_{ijk}(\mathbf r_1;\alpha,\mathbf A)
h_{lmn}(\mathbf r_1;\beta,\mathbf B)
\frac{1}{r_{12}}
h_{pqr}(\mathbf r_2;\gamma,\mathbf C)
h_{stu}(\mathbf r_2;\delta,\mathbf D).
$$

As in the calculation of the nuclear-attraction section, general ERIs are generated by differentiating the base $\langle ss|ss\rangle$ integral expression with respect to the Gaussian centers:

$$
W_{ABCD}^{ijk,lmn,pqr,stu}
=
\partial_{A_x}^i\partial_{A_y}^j\partial_{A_z}^k
\partial_{B_x}^l\partial_{B_y}^m\partial_{B_z}^n
\partial_{C_x}^p\partial_{C_y}^q\partial_{C_z}^r
\partial_{D_x}^s\partial_{D_y}^t\partial_{D_z}^u
\langle ss|ss\rangle,
$$

The Boys derivative identity remains

$$
\frac{d}{dx}F_n(x)=-F_{n+1}(x),
$$

so repeated differentiation again produces higher-order Boys functions that we have already encountered.

Cartesian ERIs are then recovered by Hermite expansion on all four centers:

$$
(ijk,lmn|pqr,stu)
=\sum c_a^{\,i}c_b^{\,j}c_c^{\,k}c_d^{\,l}c_e^{\,m}c_f^{\,n}c_g^{\,p}c_h^{\,q}c_i^{\,r}c_j^{\,s}c_k^{\,t}c_l^{\,u}
W_{ABCD}^{abc,def,ghi,jkl}.
$$


```python
import sympy as sp
alpha, beta = sp.symbols('alpha beta', real = True, positive = True)
Ax, Ay, Az = sp.symbols('Ax Ay Az', real = True)
Bx, By, Bz = sp.symbols('Bx By Bz', real = True)
gamma, delta = sp.symbols('gamma delta', real = True, positive = True)
Cx, Cy, Cz = sp.symbols('Cx Cy Cz', real = True)
Dx, Dy, Dz = sp.symbols('Dx Dy Dz', real = True)

p = alpha + beta
q = gamma + delta
pq = p + q
rho = p*q/pq

Px = (alpha*Ax + beta*Bx)/p
Py = (alpha*Ay + beta*By)/p
Pz = (alpha*Az + beta*Bz)/p

Qx = (gamma*Cx + delta*Dx)/q
Qy = (gamma*Cy + delta*Dy)/q
Qz = (gamma*Cz + delta*Dz)/q

RAB = ((Ax-Bx)**2+(Ay-By)**2+(Az-Bz)**2)
RCD = ((Cx-Dx)**2+(Cy-Dy)**2+(Cz-Dz)**2)
RPQ = ((Px-Qx)**2+(Py-Qy)**2+(Pz-Qz)**2)
```

```python
class Boys(sp.Function):
    nargs = 2

    def fdiff(self, argindex=1):
        if argindex == 2:
            return -Boys(self.args[0] + 1, self.args[1])
        raise ValueError(argindex)
```

```python
G00 = (2*sp.pi/rho)*(sp.pi/(p+q))**(3/2)*sp.exp(-(alpha*beta/p)*RAB)*sp.exp(-(gamma*delta/q)*RCD)*Boys(0, rho*RPQ)
G00
```

```python
from functools import lru_cache
@lru_cache(maxsize=None)
def get_ckn(k: int, n: int, p):
    """Coefficient C_k^n in the Cartesian-to-Hermite expansion."""
    if k < 0 or k > n:
        return sp.Integer(0)

    if n == 0:
        return sp.Integer(1) if k == 0 else sp.Integer(0)

    return get_ckn(k - 1, n - 1, p) / (2 * p) + (k + 1) * get_ckn(k + 1, n - 1, p)
```

```python
from functools import lru_cache
from itertools import product

def build_eri_derivative_table(Gp, Lmax, Avars, Bvars, Cvars, Dvars):
    Ax, Ay, Az = Avars
    Bx, By, Bz = Bvars
    Cx, Cy, Cz = Cvars
    Dx, Dy, Dz = Dvars

    all_idx = [
        (i, j, L - i - j)
        for L in range(Lmax + 1)
        for i in range(L + 1)
        for j in range(L + 1 - i)
    ]

    @lru_cache(maxsize=None)
    def G(i, j, k, l, m, n, o, p, q, r, s, t):
        if (i, j, k, l, m, n, o, p, q, r, s, t) == (0,) * 12:
            return Gp

        if i > 0: return sp.diff(G(i - 1, j, k, l, m, n, o, p, q, r, s, t), Ax)
        if j > 0: return sp.diff(G(i, j - 1, k, l, m, n, o, p, q, r, s, t), Ay)
        if k > 0: return sp.diff(G(i, j, k - 1, l, m, n, o, p, q, r, s, t), Az)

        if l > 0: return sp.diff(G(i, j, k, l - 1, m, n, o, p, q, r, s, t), Bx)
        if m > 0: return sp.diff(G(i, j, k, l, m - 1, n, o, p, q, r, s, t), By)
        if n > 0: return sp.diff(G(i, j, k, l, m, n - 1, o, p, q, r, s, t), Bz)

        if o > 0: return sp.diff(G(i, j, k, l, m, n, o - 1, p, q, r, s, t), Cx)
        if p > 0: return sp.diff(G(i, j, k, l, m, n, o, p - 1, q, r, s, t), Cy)
        if q > 0: return sp.diff(G(i, j, k, l, m, n, o, p, q - 1, r, s, t), Cz)

        if r > 0: return sp.diff(G(i, j, k, l, m, n, o, p, q, r - 1, s, t), Dx)
        if s > 0: return sp.diff(G(i, j, k, l, m, n, o, p, q, r, s - 1, t), Dy)
        return sp.diff(G(i, j, k, l, m, n, o, p, q, r, s, t - 1), Dz)

    derivatives = {}
    for a in all_idx:
        for b in all_idx:
            for c in all_idx:
                for d in all_idx:
                    key = a + b + c + d
                    derivatives[key] = G(*key)
    return derivatives
```

```python
from itertools import product

def nonzero_ckn(order, exponent):
    out = []
    for idx in range(order + 1):
        coeff = get_ckn(idx, order, exponent)
        if coeff != 0:
            out.append((idx, coeff))
    return out


def get_twoel(i, j, k, l, m, n, o, p, q, r, s, t, derivatives_eri):
    coeff_axes = (nonzero_ckn(i, alpha),nonzero_ckn(j, alpha),nonzero_ckn(k, alpha),
                  nonzero_ckn(l, beta), nonzero_ckn(m, beta), nonzero_ckn(n, beta),
                  nonzero_ckn(o, gamma),nonzero_ckn(p, gamma),nonzero_ckn(q, gamma),
                  nonzero_ckn(r, delta),nonzero_ckn(s, delta),nonzero_ckn(t, delta),
    )

    eri = sp.Integer(0)
    for term in product(*coeff_axes):
        idx_parts = []
        coeff = sp.Integer(1)

        for pair in term:
            axis_idx = pair[0]
            axis_coeff = pair[1]
            idx_parts.append(axis_idx)
            coeff *= axis_coeff

        idx = tuple(idx_parts)
        eri += coeff * derivatives_eri[idx]

    return sp.factor_terms(eri)
```

```python
def build_eri_dict(derivatives_eri):
    p_sym, q_sym, rho_sym, rAB_sym, rCD_sym, pRPQ_sym = sp.symbols("p q rho rAB rCD pRPQ", real=True)
    p_expr = alpha + beta
    q_expr = gamma + delta
    rho_expr = p_expr * q_expr / (p_expr + q_expr)
    rAB_expr = (Ax - Bx)**2 + (Ay - By)**2 + (Az - Bz)**2
    rCD_expr = (Cx - Dx)**2 + (Cy - Dy)**2 + (Cz - Dz)**2
    RPQ_expr = (
        ((alpha * Ax + beta * Bx) / p_expr - (gamma * Cx + delta * Dx) / q_expr)**2
        + ((alpha * Ay + beta * By) / p_expr - (gamma * Cy + delta * Dy) / q_expr)**2
        + ((alpha * Az + beta * Bz) / p_expr - (gamma * Cz + delta * Dz) / q_expr)**2
    )

    subs_map = {
        p_expr: p_sym,
        q_expr: q_sym,
        rho_expr: rho_sym,
        rAB_expr: rAB_sym,
        rCD_expr: rCD_sym,
        rho_expr * RPQ_expr: pRPQ_sym,
    }

    eri_dict = {}
    for key in sorted(derivatives_eri):
        expr = get_twoel(*key, derivatives_eri=derivatives_eri)
        eri_dict[key] = expr.subs(subs_map, simultaneous=True)
    return eri_dict
```

```python
eri_dict = build_eri_dict(derivatives_dict)
```

```python
# symbols already present inside the compact ERI expressions
p_sym, q_sym, rAB_sym, rCD_sym, pRPQ_sym = sp.symbols(
    "p q rAB rCD pRPQ", positive=True, real=True
)

# compact symbols
ABx, ABy, ABz, RAB2 = sp.symbols("ABx ABy ABz RAB2", real=True)
CDx, CDy, CDz, RCD2 = sp.symbols("CDx CDy CDz RCD2", real=True)

Px, Py, Pz = sp.symbols("Px Py Pz", real=True)   # bra product center
Qx, Qy, Qz = sp.symbols("Qx Qy Qz", real=True)   # ket product center
PQx, PQy, PQz = sp.symbols("PQx PQy PQz", real=True)

u = sp.symbols("u", nonnegative=True, real=True)   # Boys argument
KAB, KCD = sp.symbols("KAB KCD", positive=True, real=True)

subsdict = {
    # A-B and C-D differences
    Ax - Bx: ABx,
    Ay - By: ABy,
    Az - Bz: ABz,
    Cx - Dx: CDx,
    Cy - Dy: CDy,
    Cz - Dz: CDz,

    # compact distance symbols already used in build_eri_dict output
    rAB_sym: RAB2,
    rCD_sym: RCD2,

    # product centers P and Q
    (Ax*alpha + Bx*beta)/p_sym: Px,
    (Ay*alpha + By*beta)/p_sym: Py,
    (Az*alpha + Bz*beta)/p_sym: Pz,

    (Cx*gamma + Dx*delta)/q_sym: Qx,
    (Cy*gamma + Dy*delta)/q_sym: Qy,
    (Cz*gamma + Dz*delta)/q_sym: Qz,

    # P - Q components
    (Ax*alpha + Bx*beta)/p_sym - (Cx*gamma + Dx*delta)/q_sym: PQx,
    (Ay*alpha + By*beta)/p_sym - (Cy*gamma + Dy*delta)/q_sym: PQy,
    (Az*alpha + Bz*beta)/p_sym - (Cz*gamma + Dz*delta)/q_sym: PQz,

    -(Cx*gamma + Dx*delta)/q_sym + (Ax*alpha + Bx*beta)/p_sym: PQx,
    -(Cy*gamma + Dy*delta)/q_sym + (Ay*alpha + By*beta)/p_sym: PQy,
    -(Cz*gamma + Dz*delta)/q_sym + (Az*alpha + Bz*beta)/p_sym: PQz,

    # Boys argument
    pRPQ_sym: u,

    # Gaussian prefactors
    sp.exp(-alpha*beta*rAB_sym/p_sym): KAB,
    sp.exp(-delta*gamma*rCD_sym/q_sym): KCD,
}
```

```python
for key, value in eri_dict.items():
    eri_dict[key] = value.subs(subsdict, simultaneous=True)
```

```python
from pathlib import Path
from sympy.printing.numpy import NumPyPrinter, _known_functions_numpy, _known_constants_numpy

class TheochemNumPyPrinter(NumPyPrinter):
    def print_boys(self, expr):
        n, t = expr.args
        return f"boys({self._print(n)}, {self._print(t)})"

printer = TheochemNumPyPrinter()
printer._module = "np"
printer.known_functions = {k: f"np.{v}" for k, v in _known_functions_numpy.items()}
printer.known_functions["Boys"] = "Boys"
printer.known_constants = {k: f"np.{v}" for k, v in _known_constants_numpy.items()}
```

```python
def write_eri_module(path, name="ERI", use_cse=False, integral_expressions=None):
    idx_names = ["ia", "ja", "ka", "ib", "jb", "kb", "ic", "jc", "kc", "id", "jd", "kd"]
    param_names = [
        "Ax", "Ay", "Az",
        "Bx", "By", "Bz",
        "Cx", "Cy", "Cz",
        "Dx", "Dy", "Dz",
        "ABx", "ABy", "ABz",
        "CDx", "CDy", "CDz",
        "p", "q",
        "alpha", "beta", "gamma", "delta",
        "Px", "Py", "Pz",
        "Qx", "Qy", "Qz",
        "PQx", "PQy", "PQz",
        "rAB", "rCD", "pRPQ", "u",
        "F0", "F1", "F2", "F3", "F4",
        "KAB", "KCD",
    ]
    signature = ", ".join(["case_id"] + param_names)

    p, q = sp.symbols("p q", positive=True, real=True)
    alpha, beta, gamma, delta = sp.symbols("alpha beta gamma delta", positive=True, real=True)
    rAB, rCD, pRPQ = sp.symbols("rAB rCD pRPQ", real=True)
    KAB, KCD = sp.symbols("KAB KCD", positive=True, real=True)
    F0, F1, F2, F3, F4 = sp.symbols("F0 F1 F2 F3 F4", real=True)

    boys_subs = {
        Boys(0, pRPQ): F0,
        Boys(1, pRPQ): F1,
        Boys(2, pRPQ): F2,
        Boys(3, pRPQ): F3,
        Boys(4, pRPQ): F4,
    }

    ab_exp = -alpha * beta * rAB / p
    cd_exp = -delta * gamma * rCD / q

    def optimize_expr(expr):
        expr = sp.powsimp(sp.factor_terms(expr), force=True)
        expr = expr.replace(
            lambda node: node.func == sp.exp and sp.simplify(node.args[0] - ab_exp) == 0,
            lambda node: KAB,
        )
        expr = expr.replace(
            lambda node: node.func == sp.exp and sp.simplify(node.args[0] - cd_exp) == 0,
            lambda node: KCD,
        )
        expr = expr.replace(
            lambda node: node.func == sp.exp and sp.simplify(node.args[0] - ab_exp - cd_exp) == 0,
            lambda node: KAB * KCD,
        )
        expr = expr.subs(boys_subs, simultaneous=True)
        return sp.factor_terms(expr)

    cases = [(key, optimize_expr(value)) for key, value in integral_expressions.items()]

    lines = [
        "import numpy as np",
        "from numba import njit",
        "",
        "ERI_KEY_TO_CASE = {",
    ]
    for case_id, (key, _) in enumerate(cases):
        lines.append(f"    {key}: {case_id},")
    lines.extend([
        "}",
        "",
        "@njit(cache=True, fastmath=True)",
        f"def {name}({signature}):",
    ])

    for case_id, (_, value) in enumerate(cases):
        lines.append(f"    if case_id == {case_id}:")
        if use_cse:
            repls, reduced = sp.cse(value, symbols=sp.numbered_symbols("t"))
            for sym, expr in repls:
                lines.append(f"        {sym} = {printer.doprint(expr)}")
            lines.append(f"        return {printer.doprint(reduced[0])}")
        else:
            lines.append(f"        return {printer.doprint(value)}")

    lines.append("    raise KeyError(case_id)")

    with open(path, "w", encoding="utf-8") as f:
        f.write("\n".join(lines) + "\n")
```
