## Course Goals

- Build a compact quantum-chemical electronic-structure code from scratch.

- Understand the mathematics of Gaussian basis sets, evaluation and implementation of one- and two-electron integrals.

- Implementation of Hartree-Fock and post-Hartree-Fock methods (configuration interaction, coupled cluster, etc.) methods.

- Implement gradient-based optimization and apply it to molecular structure calculations.
