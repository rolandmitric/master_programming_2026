# The Hartree-Fock Method

 As we have learned in the basic quantum chemistry course, the electronic wavefunction of a molecular system can be approximated by a single Slater determinant composed out of molecular orbitals (MOs) $\{\chi_i\}$, which can be forther approximated as a linear combinations of atom-centered basis functions. 
A general normalized Hartree-Fock wavefunction can be written as

$$
\Psi(\mathbf r_1, \mathbf r_2, \ldots, \mathbf r_N) = \frac{1}{\sqrt{N!}} \begin{vmatrix}
\chi_1(\mathbf r_1) & \chi_2(\mathbf r_1) & \cdots & \chi_N(\mathbf r_1) \\
\chi_1(\mathbf r_2) & \chi_2(\mathbf r_2) & \cdots & \chi_N(\mathbf r_2) \\
\vdots & \vdots & \ddots & \vdots \\
\chi_1(\mathbf r_N) & \chi_2(\mathbf r_N) & \cdots & \chi_N(\mathbf r_N) \\
\end{vmatrix},
$$

The idea of the Hartree-Fock method is to determine the optimal set of MOs $\{\chi_i\}$ by minimizing the total electronic energy of the system with respect to variations in the MOs, subject to the constraint that the MOs remain orthonormal. This leads to a set of coupled integro-differential equations known as the Hartree-Fock equations, which can be solved using iterative methods to obtain the optimal MOs and the corresponding electronic energy. The energy expectation value of the Hartree-Fock wavefunction can be expressed as

$$
E_{\text{HF}} = \sum_{i=1}^N \langle \chi_i | \frac{1}{r_{12}} | \chi_i \rangle + \frac{1}{2} \sum_{i=1}^N \sum_{j=1}^N \left( \langle \chi_i \chi_j | \frac{1}{r_{12}} | \chi_i \chi_j \rangle - \langle \chi_i \chi_j | \frac{1}{r_{12}} | \chi_j \chi_i \rangle \right),
$$  

## The Hartree-Fock Equation

Using methods of variational calculus, the Hartree-Fock equations can be derived as a set of coupled integro-differential equations for the MOs:

$$
\hat F \chi_i = \epsilon_i \chi_i, \quad i = 1, 2, \ldots, N,
$$

where $\hat F$ is the Fock operator defined as 

$$
\hat F = h + \sum_{j=1}^N \left( J_j - K_j \right).
$$

with $h$ being the one-electron Hamiltonian, $J_j$ being the Coulomb operator, and $K_j$ being the exchange operator. The Coulomb and exchange operators are defined as

$$
J_j \chi_i(\mathbf r) = \int \frac{|\chi_j(\mathbf r')|^2}{|\mathbf r - \mathbf r'|} \chi_i(\mathbf r) d\mathbf r',
$$  

$$
K_j \chi_i(\mathbf r) = \int \frac{\chi_j^*(\mathbf r') \chi_i(\mathbf r')}{|\mathbf r - \mathbf r'|} \chi_j(\mathbf r) d\mathbf r'.
$$





## The Roothaan-Hall Equations

The most commonly used method for solving the Hartree-Fock equations is to expand the molecular orbitals in a finite basis of atomic orbitals. This leads to a matrix representation of the Hartree-Fock equations. A general spin-orbital can be written as a linear combination of basis functions,

$$
\chi_i = \sum_\mu C_{\mu i} \phi_\mu ,
$$

where $\phi_\mu$ are the basis functions and $C_{\mu i}$ are the corresponding expansion coefficients. Substituting this expansion into the Hartree-Fock equations and projecting onto the AO basis leads to the Roothaan-Hall equations,

$$
\mathbf F \mathbf C = \mathbf S \mathbf C \boldsymbol{\epsilon},
$$

where $\mathbf F$ is the Fock matrix, $\mathbf C$ is the matrix of molecular-orbital coefficients, $\mathbf S$ is the AO overlap matrix, and $\boldsymbol{\epsilon}$ is the diagonal matrix of orbital energies. The overlap matrix is defined as

$$
S_{\mu\nu} = \langle \phi_\mu | \phi_\nu \rangle .
$$

The matrix elements of the Fock operator are

$$
F_{\mu\nu}
=
\langle \phi_\mu | \hat F | \phi_\nu \rangle .
$$

The Fock operator contains the one-electron operator $\hat h$ and the Coulomb and exchange interactions with all occupied spin-orbitals. Therefore,

$$
F_{\mu\nu}
=
h_{\mu\nu}
+
\sum_i^{\mathrm{occ}}
\left[
(\mu\nu|ii)
-
(\mu i|i\nu)
\right],
$$

where

$$
h_{\mu\nu}
=
\langle \phi_\mu | \hat h | \phi_\nu \rangle
$$

and the two-electron integrals are defined as

$$
(\mu\nu|\lambda\sigma)
=
\int \int
\phi_\mu(1)\phi_\nu(1)
\frac{1}{r_{12}}
\phi_\lambda(2)\phi_\sigma(2)
\,d1\,d2 .
$$

The symbol $(\mu\nu|ii)$ means that the occupied spin-orbital $\chi_i$ appears in the second electron coordinate. Inserting the AO expansion

$$
\chi_i = \sum_\lambda C_{\lambda i}\phi_\lambda
$$

for the occupied spin-orbitals gives

$$
(\mu\nu|ii)
=
\sum_{\lambda\sigma}
C_{\lambda i} C_{\sigma i}
(\mu\nu|\lambda\sigma)
$$

and

$$
(\mu i|i\nu)
=
\sum_{\lambda\sigma}
C_{\lambda i} C_{\sigma i}
(\mu\sigma|\lambda\nu).
$$

Thus, the Fock matrix elements can be written explicitly in terms of the molecular-orbital coefficients as

$$
F_{\mu\nu}
=
h_{\mu\nu}
+
\sum_i^{\mathrm{occ}}
\sum_{\lambda\sigma}
C_{\lambda i} C_{\sigma i}
\left[
(\mu\nu|\lambda\sigma)
-
(\mu\sigma|\lambda\nu)
\right].
$$


At this point it is convenient to introduce the one-particle density matrix defined as

$$
P_{\lambda\sigma}
=
\sum_i^{\mathrm{occ}}
C_{\lambda i} C_{\sigma i}.
$$

where the sum runs over all occupied molecular orbitals.
With this definition, the Fock matrix becomes

$$
F_{\mu\nu}
=
h_{\mu\nu}
+
\sum_{\lambda\sigma}
P_{\lambda\sigma}
\left[
(\mu\nu|\lambda\sigma)
-
(\mu\sigma|\lambda\nu)
\right].
$$

Therefore, the Hartree-Fock equations in an AO basis take the form

$$
\sum_\nu F_{\mu\nu} C_{\nu i}
=
\epsilon_i
\sum_\nu S_{\mu\nu} C_{\nu i}.
$$

In matrix notation, this is

$$
\mathbf F \mathbf C
=
\mathbf S \mathbf C \boldsymbol{\epsilon}.
$$

Because the Fock matrix depends on the occupied molecular orbitals through the density matrix, the Roothaan-Hall equations are nonlinear. They must therefore be solved self-consistently.

## Restricted Hartree-Fock

In the following we will implement the restricted variant of the Hartree-Fock method (RHF). In RHF, each spatial molecular orbital is occupied by two electrons with opposite spin. The corresponding spin-orbitals are

$$
\chi_i^\alpha(\mathbf{x})
=
\psi_i(\mathbf{r})\alpha(\omega)
$$

and

$$
\chi_i^\beta(\mathbf{x})
=
\psi_i(\mathbf{r})\beta(\omega),
$$

where both spin-orbitals have the same spatial part $\psi_i(\mathbf{r})$. Thus, the spatial molecular orbitals are expanded as

$$
\psi_i(\mathbf{r})
=
\sum_\mu C_{\mu i}\phi_\mu(\mathbf{r}).
$$

Because every occupied spatial orbital contains one $\alpha$ and one $\beta$ electron, the spin-summed density matrix is

$$
P_{\lambda\sigma}
=
2\sum_i^{\mathrm{occ}}
C_{\lambda i}C_{\sigma i},
$$

where the sum runs over the occupied spatial orbitals. The factor of $2$ accounts for the double occupation of each spatial orbital.

Starting from the spin-orbital expression, the Coulomb contribution receives contributions from both electrons in each doubly occupied spatial orbital, whereas the exchange contribution is no longer present between electrons with opposite spin. After summing over the spin degrees of freedom, the RHF Fock matrix elements become

$$
F_{\mu\nu}
=
h_{\mu\nu}
+
\sum_{\lambda\sigma}
P_{\lambda\sigma}
\left[
(\mu\nu|\lambda\sigma)
-
\frac{1}{2}
(\mu\sigma|\lambda\nu)
\right].
$$

Here,

$$
h_{\mu\nu}
=
\langle \phi_\mu | \hat h | \phi_\nu \rangle
$$

is the one-electron core Hamiltonian matrix element, and

$$
(\mu\nu|\lambda\sigma)
=
\int\int
\phi_\mu(1)\phi_\nu(1)
\frac{1}{r_{12}}
\phi_\lambda(2)\phi_\sigma(2)
\,d1\,d2
$$

are the two-electron repulsion integrals.

The energy of the RHF wavefunction can be expressed in terms of the density matrix and the Fock matrix as

$$
E_{\text{RHF}}
=
\sum_{\mu\nu}
P_{\mu\nu}
\left[
h_{\mu\nu}
+
\frac{1}{2}F_{\mu\nu}
\right].
$$

This can be compactly written in matrix notation as

$$
E_{\text{RHF}}
=
\frac{1}{2}\mathrm{Tr}[\mathbf P (\mathbf H + \mathbf F)],
$$

which will be the expression we will use to compute the RHF energy in our implementation.


## Solving the Roothaan-Hall Equations

Since the Roothaan-Hall equations have a form of a generalized eigenvalue problem (nonlinear), it is conventient to transform it first into a standard form. This can be achieved in many different ways, but the most common method is to use canonical orthogonalization.

## Canonical orthogonalization 

Canonical orthogonalization converts the nonorthogonal atomic-orbital basis into an orthonormal basis by diagonalizing the overlap matrix. Since $S$ is Hermitian and positive definite, we can always express it in the form

$$
S = U s U^\dagger ,
$$

where $s$ is diagonal and contains the positive eigenvalues $s_i$ of $S$.

The canonical orthogonalization matrix is defined as

$$
X = U s^{-1/2}.
$$

The new basis functions are therefore

$$
\phi_i = \sum_\mu \chi_\mu X_{\mu i}
=
\sum_\mu \chi_\mu U_{\mu i}s_i^{-1/2}.
$$

We now check that this basis is orthonormal. The overlap matrix in the transformed basis is

$$
X^\dagger S X.
$$

Inserting $X=Us^{-1/2}$ gives

$$
X^\dagger S X
=
s^{-1/2}U^\dagger S U s^{-1/2}.
$$

Using

$$
U^\dagger S U = s,
$$

we obtain

$$
X^\dagger S X
=
s^{-1/2}s s^{-1/2}
=
I.
$$

Thus canonical orthogonalization produces an orthonormal basis.



To solve the Hartree--Fock equations, we write

$$
C = X C',
$$

and insert this into

$$
FC = SC\varepsilon .
$$

This gives

$$
F X C' = S X C'\varepsilon .
$$

Multiplication from the left by $X^\dagger$ gives

$$
X^\dagger F X C'
=
X^\dagger S X C'\varepsilon .
$$

Because $X^\dagger S X=I$, we obtain the ordinary eigenvalue problem

$$
F' C' = C'\varepsilon,
$$

with

$$
F' = X^\dagger F X.
$$

After diagonalizing $F'$, the molecular-orbital coefficients in the original atomic-orbital basis are recovered from

$$
C = X C'.
$$

Canonical orthogonalization is especially useful when the basis contains near-linear dependencies. If some overlap eigenvalues $s_i$ are very small, the factors $s_i^{-1/2}$ become very large and the calculation becomes numerically unstable. In practice, eigenvectors belonging to very small eigenvalues are discarded, and the Hartree--Fock problem is solved only in the remaining well-conditioned subspace.


### Initial guess

In order to start the SCF procedure, one needs an initial density matrix. The simplest choice is the core-Hamiltonian guess in which the initial density is obtained by diagonalizing the orthogonalized core Hamiltonian:

$$
h' = X^\mathrm{T} h X,
$$


and solving the eigenvalue problem. This gives the initial eigenvectors $C'$, from which one
obtains the lowest $n_{\mathrm{occ}}$ orbitals, and forms the initial occupied coefficient matrix $C_{\mathrm{occ}}$. The initial density can be constructed from the occupied orbitals as

$$
P = C_{\mathrm{occ}} C_{\mathrm{occ}}^\mathrm{T}.
$$




##  Self-consistent-field iteration

The Hartree--Fock equations are nonlinear since the Fock matrix depends on the density, and the density depends on the occupied orbitals obtained from the Fock matrix. Therefore the solution must be found by an iterative procedure.

Starting from an initial density matrix $P^{(0)}$, the iteration repeats the following steps:

1. Build the Fock matrix

   $$
   F^{(n)} = F[P^{(n)}].
   $$

2. Transform to the orthonormal basis

   $$
   F'^{(n)} = X^\mathrm{T} F^{(n)} X.
   $$

3. Diagonalize

   $$
   F'^{(n)} C'^{(n)} = C'^{(n)} \varepsilon^{(n)}.
   $$

4. Transform the coefficients back to the AO basis

   $$
   C^{(n)} = X C'^{(n)}.
   $$

5. Form the new density matrix from the occupied orbitals

   $$
   P^{(n+1)} = C_{\mathrm{occ}}^{(n)} \left(C_{\mathrm{occ}}^{(n)}\right)^\mathrm{T}.
   $$

6. Compute the electronic energy and check for convergence

   $$
   E^{(n+1)} = \mathrm{Tr}\left[\left(h+F^{(n)}\right)P^{(n+1)}\right].
   $$

   If the change in energy or density is below a certain threshold, the iteration is stopped. Otherwise, the new density $P^{(n+1)}$ is used to build a new Fock matrix and the process is repeated until convergence is achieved.  

### Convergence criteria

In our simple implementation, convergence will be declared when the change in energy between two successive iterations is smaller than a predefined threshold $\tau$:

$$
|E^{(n+1)} - E^{(n)}| < \tau.
$$

In more advanced SCF programs one often monitors additional quantities, such as the norm of the density change or the commutator error

$$
F P S - S P F,
$$

since a small energy change alone does not always guarantee that the self-consistent solution is fully stationary. Nevertheless, the energy criterion is sufficient for our basic implementation.



## Orbital occupations and returned quantities

After convergence has been achieved, the code stores the occupation numbers of the molecular orbitals in an array in which the first $$n_{\mathrm{occ}}$$ orbitals are assigned occupation 2 and all higher orbitals are assigned 0:

$$
n_i =
\begin{cases}
2, & i \le n_{\mathrm{occ}}, \\
0, & i > n_{\mathrm{occ}}.
\end{cases}
$$

The returned result object contains:

- the final electronic RHF energy,
- the AO molecular-orbital coefficient matrix $C$,
- the orbital energies,
- the orbital occupations,
- the final AO density matrix $P$ ,
- the number of SCF iterations,
- and a flag indicating whether the energy threshold was reached.


## Implementation

```python
from __future__ import annotations
from dataclasses import dataclass
import numpy as np

@dataclass
class RHFResult:
    converged: bool
    iterations: int
    energy: float
    coeff: np.ndarray
    occ: np.ndarray
    mo_energies: np.ndarray
    density: np.ndarray
```

The class `RHFResult` is a simple data class that holds the results of the RHF calculation, including the final electronic energy, the molecular-orbital coefficients, orbital energies, occupations, density matrix, number of iterations, and a convergence flag.

```python
from src_live import Molecule, BasisSet, MolecularIntegrals, ELEMENT_SYMBOL, Shell, S, T, V, ERI

class RHF:
    def __init__(self, molecule: Molecule | None = None,
                 molints: MolecularIntegrals | None = None, 
                 max_cycle: int = 100, 
                 conv_tol: float = 1e-8) -> None:
        self.molecule = molecule
        self.molints = molints  
        self.max_cycle = max_cycle
        self.conv_tol = conv_tol
        self._initialized = False

        self.nel = self.molecule.n_electrons
        self.nocc = self.nel // 2

    def initialize(self, molecule: Molecule | None = None) -> None:

        self.S = self.molints.overlap_matrix()
        self.T = self.molints.kinetic_matrix()
        self.V = self.molints.nuclear_attraction_matrix()
        self.twoel = self.molints.electron_repulsion_tensor()

        eigval, eigvec = np.linalg.eigh(self.S)
        if np.any(eigval <= 1e-12):
            raise ValueError("Overlap matrix is singular or near-singular.")
        self.X = eigvec @ np.diag(eigval ** -0.5)

        self.hcore = self.T + self.V
        hcore_ortho = self.X.T @ self.hcore @ self.X
        _, c_ortho = np.linalg.eigh(hcore_ortho)
        c_occ = self.X @ c_ortho[:, : self.nocc]
        self.P = c_occ @ c_occ.T

    def get_fock(self, P: np.ndarray) -> np.ndarray:
        g = np.einsum(
            "kl,ijkl->ij",
            P,
            2.0 * self.twoel - self.twoel.transpose(0, 2, 1, 3),
            optimize=True,
        )
        return self.hcore + g

    def run_hf(
        self,
        max_iter: int | None = None,
        threshold: float | None = None,
        verbose: bool = True,
    ) -> RHFResult:
        self.initialize()

        n_iter = self.max_cycle if max_iter is None else max_iter
        conv_tol = self.conv_tol if threshold is None else threshold
        energy_last = 0.0
        energy = 0.0
        converged = False
        iteration = 0
        mo_energies = np.array([], dtype=float)
        coeff = np.array([], dtype=float)

        for iteration in range(1, n_iter + 1):
            fock = self.get_fock(self.P)
            fock_ortho = self.X.T @ fock @ self.X
            mo_energies, c_ortho = np.linalg.eigh(fock_ortho)
            coeff = self.X @ c_ortho

            c_occ = coeff[:, : self.nocc]
            p_new = c_occ @ c_occ.T
            energy = float(np.trace((self.hcore + fock) @ p_new))

            if verbose:
                print(f"Iteration {iteration}, Energy = {energy:.12f} Hartree")

            self.P = p_new
            if abs(energy - energy_last) < conv_tol:
                converged = True
                break
            energy_last = energy

        occ = np.zeros_like(mo_energies)
        occ[: self.nocc] = 2.0

        return RHFResult(
            converged=converged,
            iterations=iteration,
            energy=energy,
            coeff=coeff,
            occ=occ,
            mo_energies=mo_energies,
            density=self.P.copy(),
        )

    def run(self, molecule: Molecule | None = None, verbose: bool = True) -> RHFResult:
        if molecule is not None:
            self.initialize(molecule)
        return self.run_hf(max_iter=self.max_cycle, threshold=self.conv_tol, verbose=verbose)
```

## Example: RHF calculation for the benzene molecule

```python
benzene_xyz = """12
Benzene molecule
C 0.000000 1.402720 0.000000
C 1.214790 0.701360 0.000000
C 1.214790 -0.701360 0.000000
C 0.000000 -1.402720 0.000000
C -1.214790 -0.701360 0.000000
C -1.214790 0.701360 0.000000
H 0.000000 2.490290 0.000000
H 2.156660 1.245150 0.000000
H 2.156660 -1.245150 0.000000
H 0.000000 -2.490290 0.000000
H -2.156660 -1.245150 0.000000
H -2.156660 1.245150 0.000000
"""
benzene = Molecule.from_string(benzene_xyz)
sto3g = BasisSet("sto-3g")
sto3g.download_from_bse(["C", "H"])
molints = MolecularIntegrals(benzene, sto3g)
rhf = RHF(molecule=benzene, molints=molints)
result = rhf.run(verbose=True)  
```
